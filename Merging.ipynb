```python
import pandas as pd
df=pd.read_csv("Datasets/all_athlete_games.csv", low_memory=False)
```

```python
#Concatenation -pd.concat()

#Concatenation is the simplest way to combine datatframes-it just stacks them either on top of each other (more rows)or side by side (more columns)
#Here there is no matching logic,no keys-it just glues things together based on position.

#axis=0 (stacking vertically,more rows)


#scenario : if we have two separate data files with summer and winter games in each,and if we have to combine them both into one dataset

summer_df=df[df['Season']=='Summer']
winter_df=df[df['Season']=='Winter']

print(summer_df.shape)
print(winter_df.shape)

combined=pd.concat([summer_df,winter_df],axis=0,ignore_index=True)
print(combined.shape)

print(combined)
```

```python
#Concatenating rows of dataframes with different columns

core_info=df[['Name','Team','Year']]
extra_info=df[['Name','Medal']]

pd.concat([core_info,extra_info],axis=0).head(50)

#When the columns don't fully match, concat (axis=0) by default keeps every column from both and fills NaN wherever a piece didn't have that column

```

```python
#Concatenating more than two pieces and tagging their source
#concat() takes a list so any number of pieces can be stacked in one call

piece1=df[df['Year']<1950]
piece2=df[(df['Year']>=1950) & (df['Year']<2000)]
piece3=df[df['Year']>=2000]

pd.concat([piece1,piece2,piece3],axis=0,ignore_index=True)
```

```python
#Tagging the Source

summer_df=df[df['Season']=='Summer']
winter_df=df[df['Season']=='Winter']

Tagged=pd.concat([summer_df,winter_df],axis=0,keys=["Summer_source","Winter_source"])
print(Tagged)
```

```python
#Stacking horizontally (More columns)
#Imagine Age came from one file and everything came from another file and both files are in the same row order

core_info=df[['Name','Team','Year']]
age_info=df[['Age']]

sideways=pd.concat([core_info,age_info],axis=1)
print(sideways)

#rows in both the files must be in the same order 
#Only then the combined dataset will be right
#axis=1 needs the same number of rows in the same order
#It does not not throw error it just takes the wrong age as it does not check the data match
```

```python
#Merging --> pd.merge()
#Merge combines two dataframes based on matching values in a shared column ("a key")
#This is like a database join
#This does not care about row order or row count it looks up the correct row to pair with
#This is based on row value 

#Basic merge --> matching on a shared column name

noc_region=pd.DataFrame({
    'NOC':['GRE','USE','FRA','GER','GBR','JPN'],
    'Continent':['Europe','North America','Europe','Europe','Europe','Asia']
})

merged=pd.merge(df,noc_region,on='NOC')
merged[['Name','NOC','Continent']]

#on 'NOC' tells pandas match rows wherever NOC values are equal in both dataframes
```

```python
#When column names don't match

noc_region_v2=pd.DataFrame({
    'country_code':['GRE','USA','FRA'],
    'Continent':['Europe','North America','Europe']
})

merged2=pd.merge(df,noc_region_v2,left_on='NOC',right_on='country_code')
merged2=merged2.drop(columns=['country_code'])
#drop the duplicate column after merge
merged2[['Name','NOC','Continent']]
```

```python
#Merging on multiple columns at once
sport_year_notes=pd.DataFrame({
    'Sport':['Athletics','Swimming'],
    'Year':[2000,2000],
    'Note':['High altitude venue','New pool record set']
})

sp=pd.merge(df,sport_year_notes,on=['Sport','Year'])
sp[['Name','Sport','Year','Note','Age','Team']]
```

```python
#merging on index instead of a column

non_region_indexed=pd.DataFrame({
    'Continent':['Europe','North America','Europe']
},
index=['GRE','USA','FRA'])

noc_in=pd.merge(df,non_region_indexed,left_on='NOC',right_index=True)
noc_in[['Name','NOC','Continent']]
#right_index=True says match on right dataframe's index not a named column
```

```python
#Suffixes, Handling overlapping non-key column names

extra_data=pd.DataFrame({
    'NOC':['GRE','USA'],
    'Age':[99,99]
})

ed=pd.merge(df,extra_data,on='NOC',suffixes=('_original','_lookup'))
ed.head()
```

```python
#Join types -inner, left, outer, right, cross
from tkinter import TRUE

from numpy import True_


noc_region_partial=pd.DataFrame({'NOC':['GRE','USA','FRA'],
                                'Continent':['Europe','North America','Europe']})

inner=pd.merge(df,noc_region_partial,on='NOC',how='inner',indicator='source')
print(inner[['Name','NOC','Continent']])
print(inner.head())
```

```python
print(inner['NOC'].unique())
print(inner.shape)
```

```python
left=pd.merge(df,noc_region_partial,on='NOC',how='left',indicator='source')
print(left['Continent'].isna().sum())
print(left['NOC'].unique())
print(left.shape)
print(left.head())
```

```python
right=pd.merge(df,noc_region_partial,on='NOC',how='right')
print(right['Continent'].isna().sum())
print(right['NOC'].unique())

outer=pd.merge(df,noc_region_partial,on='NOC',how='outer',indicator=True)
print(outer.head())
print(outer['Continent'].isna().sum())
```

```python
#Cross Join

seasons=pd.DataFrame({'Season':['Summer','Winter']})
genders=pd.DataFrame({'Gender':['Male','Female']})

Cross=pd.merge(seasons,genders,how='cross')
print(Cross)
```

```python
#.join()- shortcut for index based merging

noc_lookup=pd.DataFrame({'Continent':['Europe','North America']},index=['GRE','USA'])

df.set_index('NOC').join(noc_lookup)
```

```python
#pd.merge() - Syntaxes
pd.merge(
    left,              # first (left) DataFrame
    right,             # second (right) DataFrame
    how='inner',        # join type
    on=None,             # column(s) to match on, if the name is the SAME in both
    left_on=None,        # column(s) to match on, from the LEFT dataframe
    right_on=None,       # column(s) to match on, from the RIGHT dataframe
    left_index=False,     # match using the LEFT dataframe's index instead of a column
    right_index=False,     # match using the RIGHT dataframe's index instead of a column
    suffixes=('_x', '_y'), # tag for overlapping non-key column names
    indicator=False,        # add a column showing where each row's match came from
    validate=None            # enforce an expected key relationship, error if violated
)
```

```python
#validate - one_to_one, one_to_many, many_to_one, many_to_many

#1:1 - error if key is duplicated on either side
#1:m - error if key is duplicated on left side
#m:1 - error if key is duplicated on right side
#m:m - no uniqueness check at all (same as default behaviour even if validate is not written)

pd.merge(df,noc_region,on='NOC',validate='many_to_one')
```

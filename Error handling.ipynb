What is error handling ?

* When something goes wrong python raises an exception and stops execution.
* Error handling lets you catch that exception, do something about it, and keep the program running.

```python
#Without exception
orders = [3000,500,0,460,700]

for i in orders:
   total = 1000/i
   print(total)
```

```python
#With exception
orders = [3000,500,0,460,700]

for i in orders:
    try:
        total = 1000/i
        print(total)
    except ZeroDivisionError: #Specific error
        print("Skipping this value bcz it's a zero")
```

Rules and Properties:
1. Try block --> code that might raise an error.
2. Except block --> runs only if an error occurs in try.
3. else block --> runs only if no error occurred in try.
4. finally block --> runs always whether error occurred or not.
5. catch specific exceptions or catch all with a general except.
6. raise --> raise your own exceptions intentionally.
7. program continues after except --> exception is handled and code moves on.

### Common built-in Exceptions
| Exception | When it occurs |
|---|---|
| `ZeroDivisionError` | Division by zero |
| `TypeError` | Wrong data type used |
| `ValueError` | Right type but wrong value |
| `KeyError` | Dictionary key not found |
| `IndexError` | List index out of range |
| `FileNotFoundError` | File does not exist |
| `AttributeError` | Method does not exist on type |
| `NameError` | Variable not defined |
| `ImportError` | Module not found |

### Syntax:
######
* try:
*   #Code that might raise an error
* except SpecificError:
*   #runs if SpecificError occurs
* except AnotherError:
*   #runs if AnotherError occurs
* else:
*   #runs only if No error occurred
* finally:
*   #runs always error or not
######

```python
#Basic try-except example
d1 = {"apple":"red","bottle":"blue","phone":"pink","chilly":"red"}
k1 = ["apple","bottle","mouse","phone","chilly","desk"]

for i in k1:
    try:
        print(d1[i])
    except KeyError:
        print("This key is not found")
```

```python
#Catching multiple exceptions

orders = [{"amount":1500,"quantity":3},
          {"amount":800,"quantity":0},
          {"amount":2300,"quantity":2},
          {"product":"bottle","quantity":5}]

def  order_details(orders,index):
    try:
        order = orders[index]
        total = order["amount"]/order["quantity"]
        print("Price per item is:",total)
    except ZeroDivisionError:
        print("skipping iteam as quantity is zero")
    except IndexError:
        print("order not found at this index")
    except KeyError as k: #as k is used to store the value causing the error
        print(f"key not found {k}")

print(order_details(orders,0))
print(order_details(orders,1))
print(order_details(orders,5))
print(order_details(orders,3))
```

```python
#Catching all exceptions (General errors)

try:
    result=1000/0
except Exception as e:
    print(f"something went wrong:{e}")

#Exception catches almost everything
#Use specific exceptions when you know what might go wrong
#Use exception as a last resort safety net
```

```python
#Exception with else

order_details = {"order_id":"ORD001","amount":1000,"status":"delivered"}

try:
    status = order_details["status"]
except KeyError:
    print("key is missing")
else:
    print("status found:",status)

#else runs only when no error is found
```

```python
#with finally

order_details = {"order_id":"ORD001","amount":1000,"status":"delivered"}

try:
    status = order_details["status"]
except KeyError:
    print("key is missing")
else:
    print("status found:",status)
finally: #Works always whether error found or not
    print("All orders processed")
```

```python
#raise -> raising a custom error
def process_payment(amount):
    try:
        if amount<=0:
            raise ValueError("Amount must be greater than zero")
        else:
            print(f"Payment of Rs.{amount} must be done")
    except ValueError as e:
        print(f"Payment failed:{e}")
    finally:
        print("Payment attempt logged")

print(process_payment(500))
print(process_payment(0))
```

```python
def place_order(amount,stock):
    if stock<=0:
        raise ValueError("Cannot place order-item out of stock")
    if amount <=0:
        raise ValueError("order amount must be greater than zero")
    print(f"order placed for Rs.{amount}")

try:
    place_order(1500,0)
except ValueError as e:
    print(f"order failed:{e}")

try:
    place_order(0,1500)
except ValueError as e:
    print(f"order failed:{e}")
```

```python
#try-except-else-finally all together
def get_order_amount(orders, index):
    try:
        amount = orders[index]["amount"]
    except IndexError:
        print(f"No order at index {index}")
    except KeyError:
        print("Amount key missing")
    else:
        print(f"Amount: Rs.{amount}")
    finally:
        print("Lookup complete")

orders = [
    {"id": "ORD001", "amount": 1500},
    {"id": "ORD002","amount": 150},
    {"id": "ORD002"}
]

get_order_amount(orders, 0)
get_order_amount(orders, 1)
get_order_amount(orders, 2)
get_order_amount(orders, 5)
```

```python
#Custom Exception --> creating our own error using class keyword

class PaymentError(Exception):  #Class NameoftheError (Exception)
    pass                        #pass is do nothing

class StockError(Exception):
    pass

def place_order(amount,stock,payment_status):
    if stock<=0:
        raise StockError("Item out of stock")
    if payment_status != "paid":
        raise PaymentError("Payment not completed")
    print(f"order placed for Rs.{amount}")

try:
    place_order(1500,0,"paid")
except StockError as e:
    print(f"stock issue:{e}")


try:
    place_order(500,5,"pending")
except PaymentError as e:
    print(f"order failed:{e}")
```

| Block | When it runs |
|---|---|
| `try` | Always — contains risky code |
| `except` | Only if error occurs in try |
| `else` | Only if NO error in try |
| `finally` | Always — error or not |

# File Handling
* File handling lets you read from and write to files on your computer-csv files,text files,log files.
* In data analytics you will constantly read CSV files, write results to files, and log outputs.File handling is how python interacts with files on disk.

### Rules and Properties:

1. Always open a file before using it-using open().
2. Always close a file after use-using .close() or with satement.
3. with statement is always preferred-closes automatically even if error occurs.
4. files have modes-determines whether you read,write or append.
5. everything you read from a file is a string-convert types after reading.
6. file path matters-if file is in same folder just use filename. if elsewhere use full path.

# File modes
| Mode | Creates file? | Overwrites? | Read? | Write? | Start position |
|---|---|---|---|---|---|
| `"r"` | No — error if missing | No | Yes | No | Beginning |
| `"w"` | Yes | Yes — wipes file | No | Yes | Beginning |
| `"a"` | Yes | No — adds to end | No | Yes | End |
| `"x"` | Yes — error if exists | — | No | Yes | Beginning |
| `"r+"` | No — error if missing | No | Yes | Yes | Beginning |
| `"w+"` | Yes | Yes — wipes file | Yes | Yes | Beginning |
| `"a+"` | Yes | No | Yes | Yes | End |

"r" - Read mode only. The default mode. Opens an existing file to read it. Does not create or modify anything.

"w" - Write only. If file already exists-wipes it completely and starts fresh.

"a" - Append only. Opens a file and adds to the end. Never touches existing content.

"x" - Create only. If file already exists- throws FileExistsError.Protects you fron accidentally overwriting.

"r+" - Read and write. Opens existing file for both reading and writing. Does not wipe content. Does not create file if missing.

"w+" - Write and Read. Creates file if not exists.Wipes existing content.Allows both reading and writing.

"a+" - Append and Read. Adds to end without touching existing content. Also allows reading.



Context manager - with statement

A context manager automatically manages resources - opening and closing files - so you never have to remember to close them manually.

```python
#without with 
#manual approach

file = open("orders.txt","r")
content=file.read()
file.close()
print(content)
```

```python
#with with statement

with open("Orders.txt","r") as file:
    content=file.read()

#with - context manager automatically manages the file opened
#open - opens the file and returns a file object
#with open(filename,mode) as file  - it is returning a file object
#as file gives the object a name so you can use it inside the block
```

```python
#writing a list of lines
#writelines works on any iterable - list,tuple,dictionary,set,string

products=["Product=Earphones,Color=White,Stock=7,Brand=Apple,Pixel,Noise\n",
          "Product=Mobile,Color=Grey,Stock=77,Brand=Apple,Pixel,Samsung\n",
          "Product=Mouse,Color=Blue,Stock=8,Brand=Logitech,Arctic Fox,Asus\n"]

with open("prod","w") as f:
    f.writelines(products)

#writelines takes a list and write each line item. It does not add nextline.
#we have to include \n for next with every line item in the list 
#to get the lines in each line.
```

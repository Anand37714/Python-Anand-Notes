```python
#write a program to find the area of triangle
import math
base = float (input("Enter the base of triangle : "))
height = float (input("Enter the height of triangle : "))
area = (base * height)/2
print ("area of triangle is", area)
```

```python
#write a program to find the area of triangle using heron's formula
import math
a = float (input("Enter the value of side a: "))
b = float (input("Enter the value of side b: "))
c = float (input("Enter the value of side c: "))
s = (a + b + c)/2
area = math.sqrt (s*(s-a)*(s-b)*(s-c))
print ("area of triangle using heron's formula is", area)
```

```python
#write a program to convert temperature from celcius to fahrenheit and vice versa
# 1. Celsius to Fahrenheit
C = float(input("Enter the temperature in celsius:"))
F = (C *9/5) + 32
print("The converted value of", C, "degree celsius to fahrenheit is",F)

# 2. Fahrenheit to Celsius
F = float(input("Enter the temperature in fahrenheit:"))
C = (F-32)*5/9
print("The converted value of", F, "degree fahrenheit to celsius is",C)
```

# List Methods — Return Types

| Method | Returns | Example |
|--------|---------|---------|
| `append()` | `None` | `sales.append(100)` → None |
| `insert()` | `None` | `sales.insert(0, 100)` → None |
| `extend()` | `None` | `sales.extend([100, 200])` → None |
| `remove()` | `None` | `sales.remove(100)` → None |
| `pop()` | The removed item | `sales.pop()` → 63000 |
| `clear()` | `None` | `sales.clear()` → None |
| `index()` | Integer (position) | `sales.index(71000)` → 2 |
| `count()` | Integer (frequency) | `sales.count(71000)` → 1 |
| `sort()` | `None` | `sales.sort()` → None |
| `reverse()` | `None` | `sales.reverse()` → None |
| `copy()` | New list | `sales.copy()` → [52000, 38000...] |

---

## Key Insight

Most list methods return `None` because they modify the list **in place**.

Only these four return something useful:

| Method | What it returns |
|--------|----------------|
| `pop()` | The item that was removed |
| `index()` | Position of the value — integer |
| `count()` | How many times value appears — integer |
| `copy()` | A new independent list |

> This is why `result = sales.sort()` stores `None`.
> Sort modifies in place and returns nothing.
> Use `sorted(sales)` if you need to store the result.

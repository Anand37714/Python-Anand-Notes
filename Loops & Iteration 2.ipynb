# List comprehensions

### Rules and Prioperties

1) Always creates a new list --> The original list is never changed.
2) Square brackets[] --> same as regular list.
3) Three parts --> expression, for clause.
4) if-else goes before for --> when you want to transform every item differently.
5) if only goes after for --> when you want to filter items.


Syntax: [expression/variable, for variable in iterable, if condition]
*        [i, for i in orders, if i > 1000] 

* expression --> what goes inside the new list
* for variable in iterable --> loop over each item
* if condition --> filter --- only include if true

```python
#Add large order values into a list where order_value > 1000
#Without List Comprehension
orders = [400,6000,7000,3000,700,300,5000]
large_orders = []

for i in orders:
    if i > 1000:
        large_orders.append(i)

print(large_orders)
```

```python
#With List comprehension

orders = [400,6000,7000,3000,700,300,5000]
large_orders = [i for i in orders if i > 1000] #In place of i you can take any variable name
print(large_orders)
```

```python
#Without comprehension

orders = [400,6000,7000,3000,700,300,5000]
discounted = []

for i in orders:
    discounted.append(i*0.9)

print(discounted)

#With comprehension

discounted = [i*0.9 for i in orders]
print(discounted)
```

```python
orders = [400,6000,7000,3000,700,300,5000]
order_category = ["small" if i < 1000 else "large" for i in orders]

print(order_category)
```

```python
#create a new list with product items in upper case
products = ["laptop","phone","bag","shoes"]
products_u = [i.upper() for i in products]

print(products_u)
```

```python
status = ["delivered", "pending", "delivered", "cncelled", "pending"]
delivered = [i for i in status if i == "delivered"]

print(delivered)
```

```python
#discount to be applied only to large orders > 1000
orders = [400,6000,7000,3000,700,300,5000]

large_discounted = [i*0.9 for i in orders if i>1000]
print(large_discounted)
```

```python
#dictoaries in list

orders = [
    {"id":"ORD001","status":"delivered","amount":1500},
    {"id":"ORD002","status":"pending","amount":800},
    {"id":"ORD003","status":"delivered","amount":2300},
    {"id":"ORD004","status":"cancelled","amount":4500}
]

#Create a new list with order ids if only delivered orders

delivered_ids = [i["id"] for i in orders if i["status"] == "delivered"]
print(delivered_ids)

#create a new list with order amount of only cancelled orders

cancelled_amount = [i["amount"] for i in orders if i["status"] == "cancelled"]
print(cancelled_amount)

```

# Dictionary Comprehension

### Rules and Properties

1. Always creates a new dictionary --> original dictionary never changed.
2. Curly braces {} --> same as regular dictionary.
3. Key-Value pair - expression must have a colon seperating key and value.
4. Same filter rules as list comprehension --> if condition goes after for clause.
5. if-else goes before for --> when you want to transform every item differently.
6. if only goes after for --> when you want to filter items.



Syntax:
{Key_expression:Value_expression, for variable in iterable, if condition}

* Key_expression --> what becomes the key
* Value_expression --> what becomes the value
* for variable in iterable --> loop over each item
* if condition --> filter --- only include if true

```python
#Without comprehension 

products = {"laptop":60000,"phone":70000,"mouse":700,"shoes":10000,"marker":50}
products_discounted = {}

for i,k in products.items():
    products_discounted.update({i:k*0.9})

print(products_discounted)

#With Comprehension
products_discounted = {i:k*0.9 for i,k in products.items()}
print(products_discounted)
```

```python
products = {"laptop":60000,"phone":70000,"mouse":700,"shoes":10000,"marker":50}
products_discounted = {}

#discount with products whose value is greater than 1000
#using both without dictionary comprehension and with dictionary comprehension

#Without comprehension
for i,k in products.items():
    if k>1000:
        products_discounted.update({i:k*0.9})

print(products_discounted)
    

#With comprehension
products_discounted ={i:k*0.9 for i,k in products.items() if k>1000}
print(products_discounted)
```

```python
#create a dictionary from two lists using dict comprehension
products = ["laptop","shoes","phone","bag"]
prices = [67000,10000,65000,800]

#Normal method creating a dictionary from 2 lists
product_prices = dict(zip(products,prices))
print(product_prices)

#With comprehension creating a dictionary from 2 lists
product_prices = {i:k for i,k in zip(products,prices)}
print(product_prices)
```

```python
#Dict comprehension with filter
#Only products whose price is > 20000

product_prices = {"mobile":25000,"Laptop":85000,"Tablet":45000,"Earbuds":5000}
premium = {}

#Without comprehension
for i,k in product_prices.items():
    if k>20000:
        premium.update({i:k})

print(premium)

#With comprehension
premium = {i:k for i,k in product_prices.items() if k>20000}
print(premium)
     
```

```python
#List of dictionaries dict comprehension with filter
orders = [
    {"id":"ORD001","status":"delivered","amount":1500},
    {"id":"ORD002","status":"pending","amount":800},
    {"id":"ORD003","status":"delivered","amount":2300}
]
#a dictionary with order id as key and amount of the samre order id as value
#where status="delivered"

#with comprehension
delivered = {i["id"]:i["amount"] for i in orders if i["status"] == "delivered"} 
print(delivered)
```

```python
#Nested Loops --> Loop within another loop
categories = ["Electronics","Clothing"]
products = ["Item A","Item B","Item C"]

#Expected Output
#Electronics -  Item A
#               Item B 
#               Item C
#Clothing -     Item A
#               Item B
#               Item C

for c in categories:
    for p in products:
        print(f"category:{c}-{p}")

#To print Electronics and Clothing only 1 times

for c in categories:
    print("category",c)

    for p in products:
        print(p)
```

```python
#Nested loop for with condition
categories = ["Electronics","Clothing","Grocery"]
products = ["Laptop","Shirt","Rice","Mobile"]
electronics = ["Laptop","Mobile"]

#print category with product only if the product is an electronic item

for c in categories:
    for p in products:
        if c == "Electronics" and p in electronics:
            print("category:",c,"- product:",p)
```

```python
#Nested loop on list of dictionaries
#Expected Output
#Order :    ORD001
#           shoes
#           bag
#           watch

orders = [
    {"id":"ORD001","items":["shoes","bag","watch"]},
    {"id":"ORD002","items":["laptop","mouse"]},
    {"id":"ORD003","items":["shirt","jeans","belt"]}
]

for i in orders:
    print("Order :",i["id"])
    for k in i["items"]:
        print(f"        -{k}")
```

```python
#Expected Output
#January - North: Rs.12000
#January - South: Rs.15000
#January - East: Rs.9000
#February - North: Rs.14000
#February - South: Rs.13000
#February - East: Rs.11000
#March - North: Rs.16000
#March - South: Rs.17000
#March - East: Rs.12000

sales_grid = [
    [12000,15000,9000], #Jan
    [14000,13000,11000], #Feb
    [16000,17000,12000] #Mar
]

months = ["January","February","March"]
regions = ["North","South","East"]

for i,k in enumerate(sales_grid):
    for p,q in enumerate(sales_grid):
        print(f"{months[i]}-{regions[p]}-Rs.{sales_grid[i][p]}")
```

```python
rows = ["Row A","Row B","Row C"]
columns = ["Col 1","Col 2","Col 3"]
grid = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

#Expected Output
#Row A- Col 1:1
#Row A- Col 2:2
#Row A- Col 3:3
#Row B- Col 1:4
#Row B- Col 2:5
#Row B- Col 3:6
#Row C- Col 1:7
#Row C- Col 2:8
#Row C- Col 3:9

for i,k in enumerate(grid):
    for p,q in enumerate(grid):
        print(f"{rows[i]}-{columns[p]}:{grid[i][p]}")
```

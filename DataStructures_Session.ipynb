# List Methods — Complete Reference

## Methods Table

| Method | What it does | Syntax |
|--------|-------------|--------|
| `append()` | Adds a single item to the end | `list.append(item)` |
| `insert()` | Adds an item at a specific index | `list.insert(index, item)` |
| `extend()` | Merges another list into this list | `list.extend(another_list)` |
| `remove()` | Removes first occurrence of a value — error if not found | `list.remove(value)` |
| `pop()` | Removes and returns item at index — last item if no index given | `list.pop()` or `list.pop(index)` |
| `clear()` | Empties the entire list | `list.clear()` |
| `index()` | Returns position of first occurrence of a value | `list.index(value)` |
| `count()` | Returns how many times a value appears | `list.count(value)` |
| `sort()` | Sorts the list in place — ascending by default | `list.sort()` or `list.sort(reverse=True)` |
| `reverse()` | Flips the order of the list in place | `list.reverse()` |
| `copy()` | Returns a shallow copy of the list | `list.copy()` |

---

## Quick Examples

```python
sales = [52000, 38000, 71000, 45000, 63000]

sales.append(80000)          # [52000, 38000, 71000, 45000, 63000, 80000]
sales.insert(0, 95000)       # [95000, 52000, 38000, 71000, 45000, 63000, 80000]
sales.extend([40000, 50000]) # adds 40000 and 50000 to end
sales.remove(38000)          # removes 38000
sales.pop()                  # removes last item
sales.pop(0)                 # removes item at index 0
sales.index(71000)           # returns position of 71000
sales.count(45000)           # returns 1
sales.sort()                 # ascending
sales.sort(reverse=True)     # descending
sales.reverse()              # flips current order
sales.copy()                 # returns a copy
sales.clear()                # []

## Scenario 2 — Slicing
**Context:** You have 12 months of sales data. Pull out specific periods.

````python
monthly_sales = [42, 55, 38, 71, 63, 48, 52, 69, 74, 61, 55, 80]
#                J   F   M   A   M   J   J   A   S   O   N   D

# Qtr1 sales (Jan, Feb, Mar)
print(monthly_sales[0:3:1])


# Qtr4 sales (Oct, Nov, Dec)
print(monthly_sales[-3::1])


# First half of year
print(monthly_sales[0:6:1])


# Every alternate month
print(monthly_sales[::2])


# Reversed
print(monthly_sales[::-1])
````

###  Student Problems — Slicing

````python
scores = [55, 72, 68, 90, 45, 83, 76, 61, 88, 70]

# 1. Get the first 4 scores
print(scores[0:4])

# 2. Get the last 3 scores
print(scores[-3::])

# 3. Get scores from index 3 to 7
print(scores[3:8])

# 4. Reverse the list using slicing
print(scores[::-1])

# 5. Get every alternate score starting from index 0
print(scores[::2])
````

## Scenario 4 — Methods
**Context:** Managing a sales pipeline — adding, removing, sorting deals.

````python
pipeline = ["Deal A", "Deal B", "Deal C"]

# New deal came in
pipeline.append("Deal D")
print(pipeline)

# High priority deal — goes to top
pipeline.insert(0,"Deal E")
print(pipeline)

# Deal B fell through
pipeline.remove("Deal B")  #Using .remove, should provide value
print(pipeline)

pipeline.pop() # #Using .pop, should provide index position, by default removes last value
print(pipeline)

# Sort alphabetically
pipeline.sort()
print(pipeline)

pipeline.reverse()
print(pipeline)

pipeline.sort(reverse=True)
print(pipeline)

# How many deals?
print(len(pipeline))

# Is Deal A still in pipeline?
print("Deal A" in pipeline)

pipeline1=["a","d","b","z","y"]
print(pipeline1)

pipeline1.reverse() #Reverse as it is
print(pipeline1)

pipeline1.sort(reverse=True) #Alphabetically Reverse
print(pipeline1)



````

## Scenario 2 — Packing and Unpacking
**Context:** You receive employee data and need to work with each field separately.

````python
# Packing
employee = "Sneha", "HR", 2021, 55000
print(employee)
# Unpacking
name, dept, year, salary = employee
print(name)
print(dept)
print(year)
print(salary)
print(f"{name} joined {dept} department in {year} with salary of {salary}")
````

````python
# Partial unpacking with *
employee = "Sneha", "HR", 2021, 55000
name, *details, salary = employee
print(name)
print(details)
print(salary)
print(f"{name} Joined {details} with salary of {salary}")
````

````python
employee = ("Sneha", "HR", 2021, 55000)

name, *_, salary = employee
# Here *_ is a common convention meaning "store the unwanted values here and ignore them."

print(name)
print(_)
print(salary)
````

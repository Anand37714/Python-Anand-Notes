#Precedence and Associativity

An **expression** is anything that evaluates to a value — `2 + 3`, `x * y`, `True and x > 5`.

### Precedence Table (High → Low)

| Priority | Operator(s) | Description |
|----------|-------------|-------------|
| 1 | `()` | Parentheses |
| 2 | `**` | Exponentiation |
| 3 | `+x` `-x` | Unary sign |
| 4 | `*` `/` `//` `%` | Multiply / Divide |
| 5 | `+` `-` | Add / Subtract |
| 6 | `<<` `>>` | Bitwise shift |
| 7 | `&` `^` `\|` | Bitwise AND/XOR/OR |
| 8 | `==` `!=` `>` `<` `>=` `<=` | Comparison |
| 9 | `not` | Logical NOT |
| 10 | `and` | Logical AND |
| 11 | `or` | Logical OR |

**Associativity:** most operators → left to right · 

`**` → right to left (only exception)

```python
#Precedence decides which operator runs first
#In an expression when there are multiple operators

#Predict the output
print(3+4*2) #11
print(10/2+3) #8
print(10/(2+3)) #2
print(2**2**3) #2**8=256 (for ** --> right to left so first it will do 2**3=8 then 2**8)
print(16**2)
print(16*16)
print(15%4+2*3) #9 (% means gives the remainder-->15%4 gives the remainder as 3)
print(not 10>5 or 3<7) #not true or true --> false or true -- > true
```

```python
#Predict the output
x=5
print(x>3 and x<10 or x==7) #true and true or false --> true or false --> true
print(x>3 and (x<10 or x==7)) #true and (true or flase) --> true and true --> true
```

```python
#Predict the output
x=12
print(x>3 and x<10 or x==7) #true and false or false --> false or false --> false
print(x>3 and (x<10 or x==7)) #true and (false or flase) --> true and false --> false
```

```python
#Chained comparision
# Predict the output

marks = 73
print(90<=marks<=100) #--> 90<= marks and marks<=100 --> False and True --> False
print(70<=marks<=89) #--> 70<= marks and marks<=89 --> True and True --> True
print(50<=marks<=69) #--> 50<= marks and marks<=69 --> True and False --> False
```

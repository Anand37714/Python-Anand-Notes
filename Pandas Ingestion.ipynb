### What is Pandas?

* Pandas is a python library built on top of NumPy that provides two powerful data structures - series and DataFrame - for working with structured data
* if NumPy is for raw numbers, Pandas is for tables. Every time you work with a csv file,excel file or database in python --> Use Pandas.

### Rules and Properties:
1. Pandas is built on top of NumPy - a DataFrame is a collection of NumPy arrays, One per column, organized into a table. NumPy gives you one array. Pandas gives you many arrays - one per column - presented together as a table.
2. Two main data structures- Series(1D) and DataFrame(2D).
3. Every row and column has a label- rows have an index,columns have names.
4. Each column has its own data type- one column can be int,another float,another string,another datetime. Each column has its own NumPy array with its own type.
5. Missing values are supported-represented as NaN.

```python
#Series Vs DataFrame
#A series is a one-dimensional labelled array - like a single column from a table

import pandas as pd

ratings1=pd.Series([5.6,5.7,8.3,9.4,7.6],name="player_rating") #general indexing starting from 0
ratings2=pd.Series([5.6,5.7,8.3,9.4,7.6],name="player_rating",index=range(1,6)) #indexing from 1 to 5 using range
ratings3=pd.Series([5.6,5.7,8.3,9.4,7.6],name="player_rating",index=["P001","P002","P003","P004","P005"]) #idexing on our own
print(ratings1)
print(ratings2)
print(ratings3)
```

```python
#From a dictionary-keys automatically become index
data={"P001":5.4,"P002":6.7,"P003":8.9}
ratings=pd.Series(data)
print(ratings)
```

```python
#DataFrame
#A dataframe is a 2D labelled table-like an excel spreadsheet or a databse table
# #Each column is a series

#  small DataFrame 
data = {
    "player_name"  : ["Rodri Fati", "Ansu Le Normand", "Gavi Ramos"],
    "position"     : ["Goalkeeper", "Midfielder", "Midfielder"],
    "goals"        : [0, 0, 1],
    "player_rating": [5.6, 5.7, 8.3]
}
df=pd.DataFrame(data)
print(df)
print(type(df["goals"])) #class series bcz every column is a series
print(df.index) #range of index
print(df.columns) #column names
print(df.values) #NumPy array of all values
print(df.shape) #rows and columns in the table
print(df.dtypes) #data type of each column
```

```python
#Loading files
#Syntax for loading a csv file
#pd.read_csv(filepath,parse_dates,index_col,use_cols,nrows)

import pandas as pd
df=pd.read_csv("./Datasets/fifa_world_cup_2026_player_performance.csv")
print(type(df))
print(df.shape)
```

| Argument | Description | Optional | Default |
|---|---|---|---|
| filepath | path to file | No | — |
| parse_dates | columns to parse as dates | Yes | False |
| index_col | column to use as row index | Yes | None |
| usecols | list of columns to load | Yes | all columns |
| nrows | number of rows to load | Yes | all rows |

```python
#load with parse_dates
#Without parse_dates Pandas loads it as a plain string
#With parse_dates it loads as a proper datetimeobject you can filter and sort

#Without parse_dates
df=pd.read_csv("./Datasets/fifa_world_cup_2026_player_performance.csv")
print(df["match_date"].dtype)
print(df["match_date"][0])

#With parse_dates
df=pd.read_csv("./Datasets/fifa_world_cup_2026_player_performance.csv",parse_dates=["match_date"])
print(df["match_date"].dtype)
print(df["match_date"][0])

# Always use parse_dates for date columns only. 
# As a string you cannot filter by month or sort chronologically. 
# As datetime you can do all of that
```

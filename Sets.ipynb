# Sets

* Sets are unordered, There is no concept of index here
* Does not allow duplicates, Stores only unique values
* Can store multiple heterogeneous values
* Set is Mutable
* Stores only immutable elements --> (list, dict, set are not allowed as elements)

*Hashability

*Every value stored inside a set has a hash value

*Python stores the elements using their hash value

*Hashability is not a feature of lists or tuples. It's a property of objects that determines whether an element can be used as a dictionary key or set element

*What is a hash value?
*A hash value is a randomly generated unique integer value given to an element as id --> It's a specially generated integer value.

*Lists, dict, set are unhashable items bcz they are mutable

*Unhashable items are not allowed inside a set

*Whenever a duplocate value tries to enter the set , it is eliminated bcz of hash value 

*A unique identifier or fingerprint used by python to locate objects quickly

```python
#Creating/Defining Set

s={1,2,3,'hi'}
print(s)
print(type(s))

#Empty set
s1 = {} #Dictionary
print(s1)
print(type(s1))

s1 = set(s1) #Type casting can be used to create an empty set.
print(s1)
print(type(s1))

s3=set('Hyper')
print(s3)
print(type(s3))

s=[7,8,9]
s2=set(s)
print(s2)
print(type(s2))


s3=list(s)
print(s3)
print(type(s3))

s5={1,1,2,2,3,3,4,4,6,6} #Does not allow duplicates
print(s5)
print(type(s5))

s4 = {1,2,3,4,5.5,'hello', (1,2,3)}
print(s4)
print(type(s4))
```

```python
#add - add one item
s={1,2,3}
s.add(4)
s.add(3)
print(s)

#Update - add multiple elements
s.update((5,6,7,'hlo'))
print(s)

s.update([8,9,10,'hi'])
print(s)

#remove - remove an element (only one element can remove)
s.remove('hlo')
print(s)

#s.remove(1) #throws an error if the value is not present.
#s.remove(11)
#print(s)

#discard
s.discard(7)
print(s)

s.discard(11) #does not throw error when value is not present
print(s)
```

| Type | Allowed in Set | Reason |
|---|---|---|
| string | Yes | Fixed, cannot change |
| int, float | Yes | Fixed, cannot change |
| tuple | Yes | Fixed, cannot change |
| bool | Yes | Fixed, cannot change |
| list | No | Can change after creation |
| dict | No | Can change after creation |
| set | No | Can change after creation |

| Method | What it does |
|---|---|
| `.add()` | Adds one item |
| `.update()` | Adds multiple items |
| `.remove()` | Removes item — error if not found |
| `.discard()` | Removes item — silent if not found |
| `.pop()` | Removes a random item and returns it |
| `.difference_update()` | Removes multiple items at once |
| `.copy()` | Returns a copy of the set |
| `.clear()` | Empties the set |

| Method | Returns |
|---|---|
| `.add()` | None |
| `.update()` | None |
| `.remove()` | None |
| `.discard()` | None |
| `.pop()` | Removed item |
| `.difference_update()` | None |
| `.copy()` | New set |
| `.clear()` | None |

```python
#Set method
#add
students = {"Ganesh", "Anand", "Narmada"}
print(students)

students.add("Mahesh")
print(students)

#Update
students.update(["Shravani", "Keerti", "Kusuma", "Shravan", "Suman"])
print(students)

#remove
students.remove("Ganesh")
print(students)

#discard (safe remove option)
students.discard("Keerti")
students.discard("Anand")
print(students)

#Pop
students.pop()
print(students)

#copy
backup = students.copy()
print(backup)

#difference_update --> This works like a set operation
#through this we can remove multiple elements in the set

print(students)
students.difference_update("Kusuma", "Shravani")
print(students)
```

| Operation | Symbol | Method form | What it does |
|---|---|---|---|
| Union | `\|` | `.union()` | All items from both sets |
| Intersection | `&` | `.intersection()` | Only items present in both |
| Difference | `-` | `.difference()` | Items in one but not the other |
| Symmetric Difference | `^` | `.symmetric_difference()` | Items in either but not both (Except Only items present in both) |

```python
#Set operation
c1 = {"Anand", "Sujita", "Ganesh", "Narmada"}
c2 = {"Kusuma", "Shravani", "Deepika", "Ganesh", "Narmada"}

#Union --> Combine the 2 sets and eleminate the common items
c3 = c1.union(c2)
c4 = c1 | c2
print(c3)
print(c4)

#intersection --> It will show Only the common elements
c3 = c1.intersection(c2)
c4 = c1 & c2
print(c3)
print(c4)

#DIFFERENCE --> PRESENT IN ONE BUT NOT IN THE OTHER
c3 = c1.difference(c2)
c4 = c1 - c2 #Gives the elements present in c1 but not in c2
print(c3)
print(c4)


c3 = c2.difference(c1)
c4 = c2 - c1 #Gives the elements present in c2 but not in c1
print(c3)
print(c4)

#Symmetric difference --> elements present in either one of the set but not in both the sets.
# if the elements present in both the sets they will be eliminated
c3 = c2.symmetric_difference(c1)
c4 = c2 ^ c1
print(c3)
print(c4)

c3 = c1.symmetric_difference(c2)
c4 = c1 ^ c2
print(c3)
print(c4)
```

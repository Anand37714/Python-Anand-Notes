```python
#Control flow statements - helps the program make decisions based on conditions we give
# The conditions we give are similar to real life examples
# if sunny -> carry sunglasses
# if rainy -> carry umbrella
# if winter -> wear warm clothes

#Similarly we give conditions to the program saying:
# if this is the condition -> perform this task
# Here only if the conidition is true the code inside the block runs if not it does not
```

Rules and Properties

1. A condition always evaluates to True or False. Python decides which  block to run based on this result
2. Indentation is mandatory - the code block under every condition must be indented 4 spaces.
3. Every conditional line ends with a colon :  -missing the colon throws a syntax error.
4. Only one block runs- in an if-elif-else chain,as soon as one condition is True Python runs that block and skips everything else.
5. else is optional - if no condition matches and there is no else,nothing happens and Python moves on
6. Conditions can be combined- using and,or,not.

Truthy or Falsy statements

Python does not just check True or False. It evaluates any value as truthy or falsy. This means you can use any value directly in a condition- not just boolean True or False.

```python
#Falsy- these blocks do not run

if 0:
    print("runs")

if "":
    print("runs")

if []:
    print("runs")

if None:
    print("runs")

if False:
    print("runs")
```

```python
#Truthy - these evaluate to true

if 1:
    print("runs")

if [1,2,3]:
    print("runs")

if "paid":
    print("runs")
```

```python
#advantage of truthy and falsy- it could make our condition shorter,simpler
cart=["mouse","bottle","duster"]

if len(cart)>0:
    print("cart has items")

if cart:
    print("cart has items")
```

```python
#if-elif-else chain

product=input("Enter the category of product you want")

if product=="electronics":
    print("this is available in our store")
elif product=="grocery":
    print("this product is not available")
else:
    print("please come back next month it will be in stock")
```

```python
cost=int(input("Enter the cost of the product purchased"))

if cost>=800 and cost<1000:
    print("5% discount is applied")
elif cost>=1000 and cost<2000:
    print("10% discount is applied")
elif cost>=2000 and cost<3000:
    print("15% discount is applied")
else:
    print("Payment must be done in full price")
    
```

```python
bill_price=int(input("Enter the bill price of the customer"))

#>800
#good value

#>1500
#great value

#>2500
#amazing value

#take highest value first because order of the if-elif chain matters
if bill_price>2500:
    print("amazing value customer")
elif bill_price>1500:
    print("great value customer")
elif bill_price>800:
    print("good value customer")
```

```python
#nested if statement

#if payment is paid then print payment confirmed 
#and then check if stock is >0 then print order dispatching
#and then print thank you for shopping

payment_status="not paid"
stock=int(input("Enter the stock value available"))

if payment_status=="paid":
    print("Payment confirmed")
    if stock>0:
        print("Order dispatching")
    else:
        print("Order processing")
    print("Thank you for shopping")
```

```python
#match-case statement

#exact match
#multiple inputs that require multiple outputs

#Enter the day name and the day number of the week
day_of_the_week=input("Enter the day of the week")

print("Day entered:",day_of_the_week)

match day_of_the_week:
    case "Monday":
        print("First day of the week")
    case "Tuesday":
        print("Second day of the week")
    case "Wednesday":
        print("Third day of the week")
    case "Thursday":
        print("Fourth day of the week")
    case "Friday":
        print("Fifth day of the week")
    case "Saturday":
        print("Sixth day of the week")
    case "Sunday":
        print("Seventh day of the week")
    case _:                               #default block
        print("Invalid entry")
```

```python
#and,or,not

#if order paid and stock>0 then print order dispatched

order_status="paid"
stock=int(input("Enter the stock value available"))

if order_status=="paid" and stock>0:
    print("Order is dispatching")
else:
    print("Order is processing")
```

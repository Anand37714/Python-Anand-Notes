```python
import pandas as pd
df=pd.read_csv("Datasets/all_athlete_games.csv")
```

```python
#GroupBy - "The split, apply, combine idea"

#groupby does three steps in one linr - split, apply, and combine.
#For example if you have a dataset that you want to sort per country, then you split it first then you do some aggregation with it, count rows in it or averaging --> This is the apply part
#At the end you take all those results and stack them back into one final table --> This is the combine part.

#simple group by
df.groupby('Sport').size()

#split rows into piles by sport then count the number of rows in each pile including any missing values in other columns
```

```python
#Grouping them picking a column and applying an aggregation function to it
#df.groupby('Sport')['Age'].mean() --> result in series
df.groupby('Sport')[['Age']].mean() #dataframe 
```

```python
#groupby multiple columns 
result=df.groupby(['Sport','Season'])['Age'].mean()
result.index
df.groupby(['Sport','Season'])['Age'].mean()
#Now each pile is defined by a combination of 2 columns (sport and season) and then aggregated
            
#This is a multi-indexed result set
```

```python
#Coverting this into a normal table (flattening multi column group by back into a table)
df.groupby(['Sport','Season'])['Age'].mean().reset_index()

#reset index turns the multi-index grouping labels back into regular columns, giving you an ordinary flat table
#easier to read, filter or plot afterwards.
```

```python
#medal count by country

df.groupby(['NOC','Season'])['Medal'].count()

#Count gives you the count of rows excluding missing values
#Size gives you the count of rows including missing values
```

```python
#Aggregation --> .agg()

#We usualyy do sum,mean,count which gives us one calculation per group by.
#.agg is used when you need several calculations at once
#example --> average age and total meadl count and number of athletes, all per country in one table

#Multiple aggregation on the same column
df.groupby('NOC')['Age'].agg(['mean','min','max','count'])
```

## Numeric aggregations

| Aggregation | Meaning |
|---|---|
| `'sum'` | total of all values |
| `'mean'` | average |
| `'median'` | middle value |
| `'min'` | smallest value |
| `'max'` | largest value |
| `'std'` | standard deviation (spread around the mean) |
| `'var'` | variance |

## Counting aggregations (numeric or text)

| Aggregation | Meaning |
|---|---|
| `'count'` | number of non-missing values |
| `'size'` | number of rows total (includes missing values) |
| `'nunique'` | number of distinct (unique) values |

```python
#different aggregation for different columns
df.groupby('NOC').agg({'Age':'mean','Medal':'count','Name':'nunique'})
```

```python
#multiple Aggregations for multiple columns
df.groupby('NOC').agg({'Age':['mean','std'],'Medal':'count'})
```

```python
#naming your own aggregation columns
df.groupby('NOC').agg(avg_age=('Age','mean'),total_medals=('Medal','count'),num_athletes=('Name','nunique'))
```

```python
#Positional Aggregation

df.groupby('NOC')['Name'].first()
#This is just going to give us the first row entry

df.groupby('NOC')['Name'].last()
#This is just going to give us the last row entry
```

```python
df.sort_values('Year').groupby('Name')['Sport'].last()

#Sort the whole dataframe by year then group by athlete and last gets the Athleet's most recent sport
```

```python
#Custom functions inside aggregation
df.groupby('NOC')['Age'].agg(lambda x:x.max()-x.min())

#age range per country 
```

```python
#count how many people are having age> per country
df.groupby('NOC')['Age'].agg(lambda x:(x>50).sum())
```

```python
#named reusable custom funtion

def age_range(ages):
    return ages.max()-ages.min()

df.groupby('NOC')['Age'].agg(age_range)
```

```python
#Combining custom function with built-in function

df.groupby('NOC')['Age'].agg(['mean','std',age_range])
```

```python
#Pivoting -pivot_table, crosstab

#Both reshape our data into a grid --? rows and columns as two categories.

#Pivot_table() --> fills cells with an aggregated value you specify(mean,sum,count). More flexible

#crosstab() --> fills cells with frequency counts by default (how many rows fall into each row/column combination) 
# Simpler purpose - built for counting

df.pivot_table(index='Season',columns='Gender',values='Age',aggfunc='mean')
```

```python
df.pivot_table(index='Season',columns='Gender',values='Age',aggfunc='count')
```

```python
df.pivot_table(index='Season',columns='Gender',values='Age',aggfunc=['mean','count'])
```

```python
#Filling in missing combinations
df.pivot_table(index='NOC',columns='Season',values='Medal',aggfunc='count',fill_value=0)
```

```python
#Crosstab

pd.crosstab(df['Season'],df['Gender'])
```

```python
#normalize converts raw counts into proportions of the whole table(adding up to 1.0 overall)
#here we are normalizing across the whole table
pd.crosstab(df['Season'],df['Gender'],normalize=True)
```

```python
# normalize =index normalizes within each row,so each season's row 
# shows what fraction was male vs female adding up to 1.0 per row
pd.crosstab(df['Season'],df['Gender'],normalize='index')
```

```python
#For each country how many distinct athletes complted
df.groupby('NOC')[['Name']].agg('nunique')

#Whats the average age shown as a table with season across the rows and gender across the columns
df.pivot_table(index='Season',columns='Gender',values='Age',aggfunc='mean')

#How many entries are there for season and gender combination
df.groupby(['Season','Gender']).size() #and using crosstab we can write like this
pd.crosstab(df['Season'],df['Gender'])
```

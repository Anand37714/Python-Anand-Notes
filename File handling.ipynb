# File Handling
* File handling lets you read from and write to files on your computer-csv files,text files,log files.
* In data analytics you will constantly read CSV files, write results to files, and log outputs.File handling is how python interacts with files on disk.

### Rules and Properties:

1. Always open a file before using it-using open().
2. Always close a file after use-using .close() or with satement.
3. with statement is always preferred-closes automatically even if error occurs.
4. files have modes-determines whether you read,write or append.
5. everything you read from a file is a string-convert types after reading.
6. file path matters-if file is in same folder just use filename. if elsewhere use full path.

# File modes
| Mode | Creates file? | Overwrites? | Read? | Write? | Start position |
|---|---|---|---|---|---|
| `"r"` | No — error if missing | No | Yes | No | Beginning |
| `"w"` | Yes | Yes — wipes file | No | Yes | Beginning |
| `"a"` | Yes | No — adds to end | No | Yes | End |
| `"x"` | Yes — error if exists | — | No | Yes | Beginning |
| `"r+"` | No — error if missing | No | Yes | Yes | Beginning |
| `"w+"` | Yes | Yes — wipes file | Yes | Yes | Beginning |
| `"a+"` | Yes | No | Yes | Yes | End |

"r" = Read mode only. The default mode. Opens an existing file to read it. It does not create or modify anything.

"w" = Write mode only. If file already exists-wipes it completely and starts fresh.

"a" = Append only. Opens a file and adds to the end. Never touches existing content.

"x" = Create only. If file already exists-throws FileExistsError. Protects you from accidentally overwriting.

"r+" = Read and Write. Opens existing file for both reading and writing. It does not wipe content. Does not create file if missing.

"w+ = Write and Read. Creates file if not exists. Wipes existing content. Allows both reading and writing.

"a+" = Append and Read. Adds to end without touching existing content. Also it allows reading.

Context manager - with statement

A Context manager automatically manages resources - Opening and Closing files - so you nebver have to remember to close them manually.

```python
#without using with statement
#manual approach

file = open("Orders.txt","r")
content = file.read()
file.close()
print(content)

#if error occurs before closing the file then the file does not close
file = open("Orders.txt","r")
content = file.read()
result = 1/0
file.close()
print(content)
```

```python
#using with statement
with open("Orders.txt","r") as file:
    content = file.read()
    result = 1/0
print(content)

#with - context manager automatically manages the file opened
#open - opens the file and returns a file object
#with open(filename,mode) as file - it is returning a file object
#as file gives the object a name so you can use it inside the block
```

```python
#writing to files
#write a new file - creates file if not exists,overwrites the file if it exists
with open("Products.txt","w") as file:
    file.write("Product=Earphones, Colour=White,Stock=7,Brand=Apple,Pixel,Noise\n")
    file.write("Product=Mobile,Colour=Grey,Stock=77,Brand=Apple,Pixel,Samsung\n")
    file.write("Product=Mouse,Colour=Blue,Stock=8,Brand=Logitech,Arctic Fox,Asusy\n")
print("file created successfully")
```

```python
#wriiting a list of lines
#writelines works on any iterable - list,tuple,dictionary,set,string
products = ["Product=Earphones, Colour=White,Stock=7,Brand=Apple,Pixel,Noise\n",
            "Product=Mobile,Colour=Grey,Stock=77,Brand=Apple,Pixel,Samsung\n"
            "Product=Mouse,Colour=Blue,Stock=8,Brand=Logitech,Arctic Fox,Asusy\n"]

with open("prod","w") as file:
    file.writelines(products)

#writelines takes a list and write each line item. It does not add nextline.
#we have to include \n for next with every line item in the list to get the lines in each line.
```

```python
#list
products = ["Bottle\n","67000\n","blue\n","80\n"]

with open("prod","w") as f:
    f.writelines(products)
```

```python
#tuple
products = ("Bottle\n","67000\n","blue\n","80\n")

with open("prod","w") as f:
    f.writelines(products)
```

```python
#set
products = {"Bottle\n","67000\n","blue\n","80\n"}

with open("prod","w") as f:
    f.writelines(products)
```

```python
#set
products = {"Bottle:blue\n","Total Amount:67000\n","colour:blue\n","quantity:80\n"}

with open("prod","w") as f:
    f.writelines(products)
```

```python
#dictionary
products = {"Bottle\n":"blue\n","Total Amount\n":"67000\n","colour\n":"blue\n","quantity\n":"80\n"}
#keys will be considered , values will be ignored in dictionary

with open("prod","w") as f:
    f.writelines(products)
```

```python
#dictionary using comprehension method
products = {"Bottle":"blue\n","Total Amount":"67000\n","colour":"blue\n","quantity":"80\n"}
#To get both keys:values in dictionary 
#use comprehension method

lines=[f"{k}:{v}" for k,v in products.items()]
with open("prod","w") as f:
    f.writelines(lines)
```

```python
#Reading from files
#read entire file at once
with open("Orders.txt","r") as file:
    print(file.read())

#read returns entire file as one big string including all \n characters
```

```python
#Readlines - read all lines as a list

with open ("Orders.txt","r") as file:
    lines = file.readlines()
    print(lines)

#readlines returns a list-each line is one item including \n
```

```python
#strip to remove the \n or trailing space after every item in the list
with open ("Orders.txt","r") as file:
    lines = file.readlines()

for line in lines:
    lines=line.strip()
    print(lines)

#OR

with open ("Orders.txt","r") as file:
    lines = file.readlines()

for line in lines:
    print(line.strip())
```

```python
#read line by line using for line in file or a for loop
with open("Orders.txt","r") as file:

    for line in file:
        print(line.strip())
```

```python
#read one line at a time-readline()

with open ("Orders.txt","r") as file:
    line1 = file.readline()
    line2 = file.readline()
    line3 = file.readline()
print(repr(line1))
#repr - method to show raw representation/print along with escape sequences or hidden characters
print(line2)
print(line3)
```

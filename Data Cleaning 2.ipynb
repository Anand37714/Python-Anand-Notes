```python
#Duplicates 

#check for entire row duplicates

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv("./datasets/all_athlete_games.csv")
regions=pd.read_csv("./datasets/all_regions.csv")
df.shape

print("Full row duplicates",df.duplicated().sum())
```

```python
#exclude Entry ID -the one column auto generated that always differs

cols=[c for c in df.columns if c!='Entry ID']
print("Duplicates excluding entry ID:",df.duplicated(subset=cols).sum())
```

```python
#df.drop_duplicates()

#syntax- 
#df.drop_duplicates(subset,keep,inplace)

print("Before:",df.shape)

df_clean=df.drop_duplicates(subset=cols,keep='first')
print('After keep=first',df_clean.shape)

df_clean=df.drop_duplicates(subset=cols,keep=False)
print('After keep=False',df_clean.shape)
```

| Argument | Description | Default |
|---|---|---|
| subset | Which columns to check for duplication | None — all columns |
| keep | Which copy to keep: "first", "last", or False (drop all copies) | "first" |
| inplace | Modify original DataFrame | False |

| Method | What It Does | Example |
|---|---|---|
| .str.lower() | Converts text to lowercase | "Python" → "python" |
| .str.upper() | Converts text to uppercase | "Python" → "PYTHON" |
| .str.title() | Capitalizes the first letter of each word | "data analyst" → "Data Analyst" |
| .str.capitalize() | Capitalizes only the first letter of the string | "python" → "Python" |
| .str.strip() | Removes leading and trailing whitespace | " Python " → "Python" |
| .str.lstrip() | Removes whitespace from the left | " Python" → "Python" |
| .str.rstrip() | Removes whitespace from the right | "Python " → "Python" |
| .str.contains("text") | Returns True/False if the text appears anywhere | "Power BI" contains "BI" → True |
| .str.startswith("text") | Checks if the string starts with the given text | "Python" starts with "Py" → True |
| .str.endswith("text") | Checks if the string ends with the given text | "report.csv" ends with ".csv" → True |
| .str.replace("old", "new") | Replaces matching text | "Power BI" → "Power Query" |
| .str.split(",") | Splits a string into a list | "A,B,C" → ["A", "B", "C"] |
| .str.len() | Returns the length of each string | "Python" → 6 |
| .str.count("text") | Counts occurrences of a substring | "banana" count "a" → 3 |
| .str.find("text") | Returns the index of the first occurrence (-1 if not found) | "Python" find "th" → 2 |
| .str.slice(0, 3) | Extracts characters by position | "Python" → "Pyt" |
| .str.isalpha() | Checks if all characters are letters | "Python" → True |
| .str.isnumeric() | Checks if all characters are digits | "12345" → True |
| .str.isalnum() | Checks if all characters are letters or digits |

```python
#Regex -basic patterns

# Regex (regular expressions) is a mini pattern-matching language for text — 
# instead of searching for an exact word,
# you describe a shape of text you're looking for.
```

| Symbol | Meaning | Example |
|---|---|---|
| \D | Any non-digit character | letter, other characters |
| \d | Any single digit (0–9) | \d\d\d\d matches a 4-digit year |
| \w | Any letter, digit, or underscore | \w+ matches a whole word |
| ^ | Start of the text | ^A matches text that begins with A |
| $ | End of the text | son$ matches text that ends with "son" |
| + | One or more thing before it | \d+ matches 5,42,1500- any run of digits |
|(...)| a "Capture group"- marks the part you actually wanted to extract | -(\d)$ matches the whole -3 but only captures the 3 |

* subset=[...] → only these columns decide if two rows are "the same"
* keep=False → flag every row in a duplicate group, not just the extras, so you can inspect them

* keep='first' keeps the first occurrence and drops the rest. keep='last' keeps the most recent instead. 
* Keep=False drops all copies if you don't trust any of them.

* drop_duplicates(subset=['Name']) alone would delete an athlete's second sport entirely Choosing the right subset is the most important decision here.

```python
dup_mask=df.duplicated(subset=['Name','Year','Event'],keep=False) #Extract duplicates based on subset of columns
df[dup_mask].sort_values(['Name','Year']).head(10) #Dataframe of duplicates
```

```python
team1=pd.Series(['Greece','Great Britain','New Greece'])
print(team1.str.contains(r'^Greece')) #Starts with Greece

team2=pd.Series(['Greece','Great Britain','1-Greece'])
print(team2.str.contains(r'\d$')) #Ends with a digit

team3=pd.Series(['Greece-111','Great Britain','1-Greece','Greece-46','Greece-1'])
print(team3.str.contains(r'\d+$')) #Ends with one or more digits

team4=pd.Series(['Greece1','Great Britain','123','G1'])
print(team4.str.contains(r'\D')) #non-digit

team5=pd.Series(['Greece-1','Greece-12','Greece@1','Greece&&12','Greece&&1'])
print(team5.str.contains(r'\D\d$')) #One non digit before ending with exactly one digit

```

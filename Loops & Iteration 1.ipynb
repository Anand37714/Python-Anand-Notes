Loops & Iteration:

what are Loops?
In real life we do repititive tasks all the time:

1. Check every items in a cart and calculate total
2. Go through every order and find delivered ones
3. Keep asking for valid password until the user enters the correct one

Without loops you would have to write the same code over and over:
print(cart[0])
print(cart[1])
print(cart[2])
print(cart[3]) and so on...
What if cart has 500 items? You would need 500 lines of code

With a loop:
for items in cart:
    print(items)
Same result. One line of code instead of 500. And this works whether the cart has 5 items or 5000 items also.

Before writing a loop we need to know what all it needs:
A loop needs for keyword, iterable, in keyword, sequence

An iterable is anything that can give you one item at a time.
Think of it like a cashier at a billing queue, he attends to one consumer at a time.
Each person here is an item and the queue of people is the sequence containing myltiple items.

cat = ["shoes","bag","laptop"]
when python sees for item in cart, it asks cart - "give me your items one by one".
Cart listens and hands over shoes first, then bag, then laptop.

What makes something iterable?
1) It has a defined order or sequence of items.
2) Pyhtin can ask it for the next item.
3) It knows when it has run out of the items

| Iterable | Why it works |
|---|---|
| List `[1, 2, 3]` | Has items in order |
| String `"hello"` | Has characters in order |
| Dictionary `{"a": 1}` | Has keys in order |
| Tuple `(1, 2, 3)` | Has items in order |
| Set `{1, 2, 3}` | Has items — no guaranteed order |
| Range `range(5)` | Generates numbers on demand |

```python
#Iteration over a list
cart = ["phone", "laptop", "bag", "shoes"]

#for items in sequence: [syntax]
for items in cart:
    print(items)

#The loop variable item is not fixed. It changes every single iteration.
#Python replaces the value automatically

#Variable name here does not matter, you choose it.
#Choose something meaningful
```

```python
#Iteration over a string

fruit = "mango"

for character in fruit:
    print(character)
```

```python
#Iteration over a set

s1 = {"mango", "apple", "grapes"}

for fruit in s1:
    print(fruit)
```

```python
#Iteration over a dictionary

d1 = {"Apple":"Red", "Banana":"Yellow", "Grapes":"Green"}

for items in d1: #By default it will print the keys
    print(items)

for items in d1.values(): #If you want values inside the keys
    print(items)

for key,value in d1.items(): #If you want key:value
    print(f"{key}:{value}")

for pair in d1.items(): #If you want pair together in a tuple
    print(pair)
```

```python
#Iteration over a range:
#range(start,stop,step)

#print numbers from 0 to 4
for i in range(5):
    print(i)
#print numbers from 1 to 5
for i in range(1,6):
    print(i)

#print numbers from 1 to 5 in reverse order
for i in range(6,0,-1):
    print(i)
```

```python
#for loop by default only give the values
#enumerate - to print index along with value

cart = ["apple", "mouse", "shoes", "laptop", "bag"]

for index,item in enumerate(cart): #you can give any varianble name in yhe place of index and item like x,y
    print(f"{index}:{item}")
    
for index,item in enumerate(cart): #you can give any varianble name in yhe place of index and item like x,y
    print(f"{index+1}:{item}")
```

```python
#Condition along with loop

bill_value = [100,900,1000,5000,10000,6000,90,5600]

for i in bill_value:
    if i>1000:
        print(i)
```

# Rules and Properties
1. Loops repeat a block of code — the block runs once for each item in a sequence or until a condition becomes False.
2. Indentation is mandatory — everything inside the loop must be indented 4 spaces.
3. Every loop line ends with a colon : — missing it throws a SyntaxError.
4. Two types of loops — for iterates over a sequence. while runs until a condition is False.
5. Loops can be nested — a loop inside a loop.
6. Three control keywords — break exits the loop, continue skips to next iteration, pass does nothing.

```python
#list of orders
#PRINT IT IN THIS FORMAT --> Order 1:ORD001

orders = ["ORD001", "ORD002", "ORD003", "ORD004"]

for i in range(len(orders)): #It can print if the extra values added in orders
    print(f"Order{i+1}:{orders[i]}")

for i in range(3): #It is fixed
    print(f"Order{i+1}:{orders[i]}")
```

```python
#List of dictionaries
#PRINT EVERY ORDER IN THIS FORMAT --> ORD001-Delivered - RS.1500
orders = [
    {"id":"ORD001", "status":"delivered", "amount":1500},
    {"id":"ORD002", "status":"pending", "amount":800},
    {"id":"ORD003", "status":"delivered", "amount":2300}
]

for i in orders:
    print(f"{i["id"]}-{i["status"]}-{i["amount"]}")
```

```python
#using conditional statement

for i in orders:
    if i["status"] == "delivered":
        print(f"{i["id"]}-{i["status"]}-{i["amount"]}")
```

#While loops:

A while loop keeps running as long as the condition is true. 
It does not iterate over a sequence - it checks condition before every single iteration.

Syntax:
while condition:
    #code to run while condition is true

A while loop is used when you don't know how many times you need to repeat.
It depends on the condition, not a fixed count.

```python
payment_status = input("Enter the payment status")

while payment_status != "paid":
    print("Enter the payment_status")
```

```python
#While loop using break statement

payment_status = input("Enter the payment status")

while payment_status == "paid": #performs the task until the condition is true
    print("Payment was done")
    break
```

```python
stock = 5

while stock > 0:
    print(f"Selling stock - stock left:{stock}")
    stock = stock - 1
```

```python
#for loop using Continue (continue and break both can use in for loop)

bill_value = [500,600,0,700,590,0,0,700]

for i in bill_value:
    if i == 0:
        continue #it skips the current iteration
    print("The bill value is : ",i)
```

```python
#While loop using continue

bill_value = [500,600,0,700,590,0,0,700]
index = 0
while index < (len(bill_value)):
    if bill_value[index] == 0:
        index = index + 1
        continue #It skips the current iteration
    print("The bill value is :",bill_value[index])
    index = index + 1
```

```python
#While loop using Pass

bill_value = [500,600,0,700,590,0,0,700]
index = 0
while index < (len(bill_value)):
    if bill_value[index] == 0:
        index = index + 1
        pass
    print(f"Theis is {index}th iteration : The bill value is :",bill_value[index])
    index = index + 1
```

```python
#pass --> do nothing / place holder

categories = ["electronics", "grocery", "stationary", "furniture", "grocery"]

for i in categories:
    if i == "grocery":
        pass
    else:
        print("The category is : ",i)
```

```python
orders = [
    {"id":"ORD001","status":"delivered","amount":1500},
    {"id":"ORD002","status":"pending","amount":0},
    {"id":"ORD003","status":"cancelled","amount":800},
    {"id":"ORD004","status":"delivered","amount":2300},
    {"id":"ORD005","status":"delivered","amount":300},
]

#skip orders with amount 0 using continue
#stop at cancelled using break
#print only valid orders that are delivered

#expected output ORD001 - Delivered - Rs.1500

for i in orders:
    if i["amount"] == 0:
        continue #it skips the current iteration
    elif i["status"] == "cancelled":
        continue #it skips the current iteration
    elif i["status"] == "delivered":
        print(f"{i["id"]}, {i["status"]}, Rs.{i["amount"]}")
```

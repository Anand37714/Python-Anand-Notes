## Section 3 — Set Operations

### Question 4
Compare morning and evening shift doctors.

```python
morning = {"Dr. Sharma", "Dr. Patel", "Dr. Rao", "Dr. Nair"}
evening = {"Dr. Rao", "Dr. Nair", "Dr. Mehta", "Dr. Khan"}
```

```python
# a. Which doctors worked both shifts?
both_shifts1 = morning.intersection(evening)
both_shifts2 = morning & evening
print(both_shifts1)
print(both_shifts2)
```

```python
# b. All doctors who worked today
All_Doctors1 = morning.union(evening)
All_Doctors2 = morning | evening
print(All_Doctors1)
print(All_Doctors2)
```

```python
# c. Doctors who worked only the morning shift
only_mrng_shift1 = morning.difference(evening)
only_mrng_shift2 = morning - evening
print(only_mrng_shift1)
print(only_mrng_shift2)
```

```python
# d. Doctors who worked only the evening shift
only_evng_shift1 = evening.difference(morning)
only_evng_shift2 = evening - morning
print(only_evng_shift1)
print(only_evng_shift2)
```

```python
# e. Doctors who worked exactly one shift
worked_exactly_one_shift1 = morning.symmetric_difference(evening)
worked_exactly_one_shift2 = morning ^ evening
print(worked_exactly_one_shift1)
print(worked_exactly_one_shift2)
```

### Question 5
Three wards logged their admitted patients.

```python
ward_a = {"P001", "P002", "P003", "P004"}
ward_b = {"P003", "P004", "P005", "P006"}
ward_c = {"P005", "P006", "P007", "P008"}
```

```python
# a. Patients common to ward A and ward B
common_patients1 = ward_a.intersection(ward_b)
common_patients2 = ward_a & ward_b
print(common_patients1)
print(common_patients2)
```

```python
# b. Patients common to all three wards
common_patients_to_all_3_wards1 = ward_a.intersection(ward_b).intersection(ward_c)
common_patients_to_all_3_wards2 = ward_a & ward_b & ward_c
print(common_patients_to_all_3_wards1)
print(common_patients_to_all_3_wards2)
```

```python
# c. All unique patients across all three wards
all_unique_patients1 = ward_a.union(ward_b).union(ward_c)
all_unique_patients2 = ward_a | ward_b | ward_c
print(all_unique_patients1)
print(all_unique_patients2)
```

```python
# d. Patients in ward A but not in ward B or C
Patients_in_ward_A_but_not_in_ward_B_or_C1 = ward_a.difference(ward_b).difference(ward_c)
Patients_in_ward_A_but_not_in_ward_B_or_C2 = ward_a - ward_b - ward_c
print(Patients_in_ward_A_but_not_in_ward_B_or_C1)
print(Patients_in_ward_A_but_not_in_ward_B_or_C2)
```

```python
# e. Patients in ward B and C but not in ward A
Patients_in_ward_B_and_C_but_not_in_ward_A1 = ward_b.intersection(ward_c).difference(ward_a)
Patients_in_ward_B_and_C_but_not_in_ward_A2 = ward_b & ward_c - ward_a
print(Patients_in_ward_B_and_C_but_not_in_ward_A1)
print(Patients_in_ward_B_and_C_but_not_in_ward_A2)
```

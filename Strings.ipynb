String methods-
| Method | What it does |
|---|---|
| `.upper()` | Converts to uppercase |
| `.lower()` | Converts to lowercase |
| `.title()` | First letter of each word capitalised |
| `.strip()` | Removes leading and trailing spaces |
| `.lstrip()` | Removes leading spaces only |
| `.rstrip()` | Removes trailing spaces only |
| `.replace()` | Replaces one substring with another |
| `.split()` | Splits string into a list |
| `.join()` | Joins a list into a string |
| `.find()` | Returns index of first match — -1 if not found |
| `.index()` | Returns index of first match — error if not found |
| `.count()` | Counts how many times substring appears |
| `.startswith()` | True if string starts with given value |
| `.endswith()` | True if string ends with given value |
| `.isdigit()` | True if all characters are digits |
| `.isalpha()` | True if all characters are letters |
| `.isupper()` | True if all characters are uppercase |
| `.islower()` | True if all characters are lowercase |
| `.zfill()` | Pads string with zeros on the left |
| `.center()` | Centers string with padding character |
| `.format()` | Inserts values into a string template |

# String Methods — Complete Reference

| Method | Syntax | Required Args | Optional Args | Returns |
|---|---|---|---|---|
| `.upper()` | `str.upper()` | None | None | New string |
| `.lower()` | `str.lower()` | None | None | New string |
| `.title()` | `str.title()` | None | None | New string |
| `.isupper()` | `str.isupper()` | None | None | Boolean |
| `.islower()` | `str.islower()` | None | None | Boolean |
| `.isdigit()` | `str.isdigit()` | None | None | Boolean |
| `.isalpha()` | `str.isalpha()` | None | None | Boolean |
| `.strip()` | `str.strip(chars='')` | None | chars | New string |
| `.lstrip()` | `str.lstrip(chars='')` | None | chars | New string |
| `.rstrip()` | `str.rstrip(chars='')` | None | chars | New string |
| `.replace()` | `str.replace(old, new, count=-1)` | old, new | count | New string |
| `.split()` | `str.split(sep=None, maxsplit=-1)` | None | sep, maxsplit | List |
| `.join()` | `str.join(iterable)` | iterable | None | New string |
| `.find()` | `str.find(sub, start=0, end=len(str))` | sub | start, end | Integer or -1 |
| `.index()` | `str.index(sub, start=0, end=len(str))` | sub | start, end | Integer or ValueError |
| `.count()` | `str.count(sub, start=0, end=len(str))` | sub | start, end | Integer |
| `.startswith()` | `str.startswith(prefix, start=0, end=len(str))` | prefix | start, end | Boolean |
| `.endswith()` | `str.endswith(suffix, start=0, end=len(str))` | suffix | start, end | Boolean |
| `.zfill()` | `str.zfill(width)` | width | None | New string |
| `.center()` | `str.center(width, fillchar=' ')` | width | fillchar | New string |
| `.format()` | `str.format(*args, **kwargs)` | None | any args | New string |

```python
#Upper, Lower, Title, Strip, lstrip, rstrip
name="|     Mr.sharma|"
print(name.upper())
print(name.lower())
print(name.title()) #After every word it will convert the first letter as capital

name="     Mr.sharma is     not    well"
print(name.strip())
print(name.rstrip())
print(name.lstrip())
print(name.title().strip())
```

```python
#Replace, Split, Join, Find, Index
#Replace --> Accepts three arguments (old,new,count)
statement1= "Red was the colour of the Apple"
statement1.replace("Red","green")
print(statement1)
                  
statement2 = "Red was the colour of the Apple and Red was also used in the interiors"
statement2.replace("Red","green")
statement2.replace("Red","Green",1)
print(statement2)

#Split --> A list after splitting the string
Fruits = "Apple,Banana,Mango,Pear"
f1 = Fruits.split(",")
print(f1)

Fruits = "Apple Banana Mango Pear"
f1 = Fruits.split()
print(f1)

#Join
Fruits = ["Apple","Banana","Mango"]
s1 = " ".join(Fruits) #Any character can be used within quotes to create a string
print(s1)

#Find, Index
#Find (Values to find, start, stop)
#Find --> Finds the position at which the character is present
statement3 = "Red was the color of the Apple and Red was also used in the interiors"
print(statement3.find("r"))
print(statement3.find("in"))
print(statement3.find("color"))
print(statement3.find("is")) #-1 for a value not present
print(statement3.find("interiors"))

print(statement3.find("the",10,25))
print(statement3.find("was"))
print(statement3.find("was",7))
print(statement3.find("was",5,10))

#index
print(statement3.index("r"))
print(statement3.index("in"))
print(statement3.index("color"))
print(statement3.index("interiors"))


print(statement3.index("the",10,25))
print(statement3.index("was"))
print(statement3.index("was",7))
print(statement3.index("was",5,10))
```

```python
#count, startwith, endswith

statement2 = "Red was the colour of the Apple and Red was also used in the interiors"
print(statement2.count("the"))
print(statement2.count("o"))

print(statement2.startswith("R"))
print(statement2.startswith("Red"))
print(statement2.startswith("blue"))
print("blue lagoon lake".startswith("blue"))

print(statement2.endswith("s"))
print(statement2.endswith("interiors"))
print(statement2.endswith("blue"))
print("blue lagoon lake".endswith("lake"))

a="blue lagoon lake"
a[-1:-5:-1]
```

```python
#isdigit, isnumeric, isdecimal, isalpha, isupper, islower, isalnum

a="123"
print(a.isdigit())
print(a.isnumeric())
print(a.isdecimal())

b="abc"
print(b.isalpha())

c="abc123"
print(c.isdigit())
print(c.isalpha())
print(c.isalnum())

d="HELLO"
print(d.islower())
print(d.isupper())
```

```python
#Zfill,Center,Format

name = "|Sam|"

#Zfill --> is to pad zeroes on the left side
print(name.zfill(10)) #By default string will allign at right and with z fill 0 will fill at left side by alligning the string at right side by default

#Center --> for center allignment
print(name.center(10))
print(name.center(10,'*'))

#format --> positional arguments
print("{} was the colour of the {} and {} was also used in the interiors".format("Red","Apples","Red"))
print("{} {} were {} in quantity".format("Red","Apples",10))
print("{} coloured {} were {} in quantity".format("Red","Apples",10))
print("My name is {}".format("Anand"))

#format --> keyword arguments
print("{color} {fruit} were {quantity} in quantity".format(quantity = 10, color = "Yellow", fruit = "Mango"))
```

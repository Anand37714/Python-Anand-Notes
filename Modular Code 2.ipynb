# *args --> Rules and Properties

1. *args collects all extra positional arguments into a tuple.
2. The name args is convention --> you can use any name after *.
3. *args must come after regular positional parameters.
4. you can loop over *args --> it is tuple.
5. only one *args can be used per function

```python
def outer(x):
    def inner(y):
        return x + y
    return inner

add_five = outer(5)
print(add_five(10))
```

```python
#Function definiyion with args parameter
#Used when we don't know how many positional arguments will be passed
def total_bill(*bill_amounts):
    print(type(bill_amounts))
    print(bill_amounts)
    return sum(bill_amounts)

print(total_bill(670,780,450,700))
print(total_bill(12,45,90))
```

```python
def create_order(status,*order_items):
    print(type(order_items))
    print(status)
    for order in order_items:
        print("order:",order)
    return "orders printed"

print(create_order("delivered","shoe","bag","bottle"))
print(create_order("pending","jersey","pen","paint","keyboard","mouse"))
```

```python
#create a function taking multiple order_ids, print the type of argument
#also print the order id one by one using loop (take atleast 5 arguments)
#output -   Processing:ORD001
#           Processing:ORD002

def orders(*order_ids):
    print(type(order_ids))
    for i in order_ids:
        print("Processing:",i)
    return "Orders Processed"

print(orders("ORD001","ORD002","ORD003","ORD004","ORD005"))
print(orders("ORD006","ORD007","ORD008","ORD009","ORD0010"))
```

```python
#Take multiple order amounts and print the total along with order amounts

#Output:
#   orders:(1500,700)
#   Total:Rs.2200
#   orders:(1500,)
#   Total:Rs.1500

def orders(*bill_amounts):
    Total=sum(bill_amounts)
    print("orders:",bill_amounts)
    print("Total: Rs.",Total)
    return "orders closed"

print(orders(1500,700))
print(orders(1500,))
```

## **Kwargs --> Let you pass any number of keyword arguments. pyhton collects them into dictionary
### Rules and Properties:
1. **kwargs collects all extra keyword arguments into a dictionary
2. The name kwargs is convention --> you can use any name after **
3. **kwargs must come after regular parameters and *args
4. keys are strings --> the argument names become dictionary keys
5. only one **kwargs can be used per function

```python
# *Kwargs (KW=key word)
def order_details(**ord):
    print(type(ord))
    for k,v in ord.items():
        print(f"{k}:{v}")

order_details(product="Bottle",quantity="7",stock=10,color="red")
```

```python
def create_order(order_id,*items,**details):
    print("Order:",order_id)
    print("Items:",items)
    print("Details:",details)
    return "Items Details"
print(create_order("ORD001","Shoe","bag","bottle",city="Banglore",status="Pending",payment="card"))
print(create_order("ORD002","Shoe","bag","pen","keyboard",city="Manglore",status="pending",payment="upi"))
```

# Rules and Properties of Lambda
1. Written in one line --> cannot span multiple lines'
2. No def, No return required here --> the expression is automatically returned
3. can have any nyumber of parameters-including none type also
4. only one expression --> cannot contain statements, loops, or conditions
5. often used inline-passed directly to another function

```python
#Lambda --> Lambda is an anonymous function without a name, written in one line.

#Regular function:
def double(x):
    return x*2

print(double(2))

#Lambda function
#Syntax --> lambda parameter:expression 
double = lambda x:x*2
Square = lambda y:y**2
print(double(2))
print(Square(5))

```

```python
#Basic lambda examples:

#with no parameter
greet = lambda:"hello"
print(greet())

#with one parameter
cube = lambda z:z**3
print(cube(5))

#with two or more parameters
discount = lambda amount,discount : amount-(amount*discount)
print(discount(400,0.05)) #positional arguments

#with two or more parameters for keyword arguments
order_details = lambda product,quantity,amount : f"The quantity purchase is:{quantity}, total cost:{amount}, product purchased:{product}"
print(order_details(quantity=4,product="Markers",amount=400)) #Keyword arguments
```

```python
#with condition
print("with condition")
get_delivery = lambda amount:"free delivery" if amount>1000 else "Rs.50/- will be charged"
print(get_delivery(500))
print(get_delivery(1500))

#with default value
get_delivery = lambda amount,threshold=1000:"free delivery" if amount>threshold else "Rs.50/- will be charged"
print("with default value")
print(get_delivery(1500))
print(get_delivery(500))

#with args
print("with args")
order_total = lambda *order_amounts:sum(order_amounts)
print(order_total(500,700,800))
print(order_total(500,700,800,900,1000))

#with kwargs
print("with kwargs")
order_details = lambda **details:f"The product details are:{details}"
print(order_details(product="Marker",amount=50,quantity=1,city="Banglore"))
```

```python
## Functional Tools

### Map, Filter, Reduce

#Map --> Map() applies function to every item in the iterable and returns the result.
#       Syntax: map(function,iterable)

#Filter --> filter() applies a function to every item and keeps only the items where the function is true.
#           Syntax: filter(function,iterable)

#Reduce --> reduce() applies a function cumulatively to reduce a list into a single value.
#           Syntax: reduce(function,iterable)
```

```python
#map
order_amounts = [700,600,660,7500]

discount = list(map(lambda amount:amount*0.9,order_amounts))
discounts = tuple(map(lambda amount:amount*0.9,order_amounts))
discounted = set(map(lambda amount:amount*0.9,order_amounts))

print(discount)
print(discounts)
print(discounted)
```

```python
names = ["priya","raahul","sanjana","anand","nagu"]
#using map and lambda convert every name into upper case

upper_case_names = list(map(lambda name:name.upper(),names))
print(upper_case_names)
```

```python
words = ["rat","singer","music","range"]
#usinf map find length of each word

def length(name):
    return len(name)

words_length = list(map(len,words))
print(words_length)

#using map and lambda finding the length of each word
words_lengths = list(map(lambda word:len(word),words))
print(words_lengths)

#using directly map function
words_length = list(map(len,words))
print(words_length)
```

```python
#Reduce

from functools import reduce
order_amounts = [700,600,660,7500,6000,7500,1600]
#sum of all the order amounts

#Reduce --> reduce() applies a function cumulatively to reduce a list into a single value.
summation = reduce(lambda a,b:a+b,order_amounts)
print(summation)
```

```python
#Reduce to find maximum value in a list
order_amounts = [700,600,660,7500,6000,7500,1600]

Maximum = reduce(lambda a,b:a if a>b else b,order_amounts)
print(Maximum)
```

```python
#Reduce to find maximum lettered word in a list
words = ["cat","elephant","giraffe","dog","air"]

maximum = reduce(lambda a,b: a if len(a)>len(b) else b,words)
print(maximum)
```

```python
k = [[6,7],[8,10],[7,10,5,6],[2,3,4]]
#convert this structure into one single list with all numbers inside it
#using reduce
#output --> [6,7,8,10,7,10,5,6,2,3,4]

from functools import reduce

concat = tuple(reduce(lambda x,y:x+y,k))
print(concat)
```

```python
#[6,5,8,10,2]
#find which number has the highest power using reduce

i = [6,5,8,10,2]
highest_power = reduce(lambda a,b: a if a**2 > b**2 else b,i)
print(highest_power)

import math
highest_power = reduce(lambda a,b:a if math.pow(a,2) > math.pow(b,2) else b,i)
print(highest_power)
```

| Letter | Scope | Where |
|-|-|-|
| L | Local | Inside the current function |
| E | Enclosing | Inside any enclosing function (outer function) |
| G | Global | At the top level of the file |
| B | Built-in | Python's built-in names like print, len, range |

* Scope
* Scope determines where a variable can be accessed. Not every variable is visible everywhere in your code.

* A variable defined inside the function cannot be seen outside it.
* A variable defined outside the function can be seen inside unless a local one with the same exist.


```python
#Scope of a variable using LEGB Rule --> Local,Enclosing,Global,Built-In

r=6  #--> Global variable

def add(a,b):
    return a+b+r #--> Local variable

Total = add(6,7)
print(Total)
print(r) #--> Global variable can be accessed outside and inside the function
print(a,b) #--> Enclosed variable can be accessed outside the function if it only exist inside
```

```python
#Local scope

def get_discount():
    discount=10
    print("Discount added:",discount,"%")

get_discount()
print(discount) #--> here discount is a local variable and it cannot be accesed outside the function
```

```python
#Global scope

name = "quick start"

def show_store():
    print(f"store name is:",name)

show_store()
print(name) #--> here name is a global variable so it can be accessed inside and outside the function
```

```python
#Local variable shadows the global variable
#This is when we have same variable name at both places

discount=13 #--> Global variable

def get_discount():
    discount=10 #--> Local variable
    print("Discoun added:",discount,"%")

get_discount()
print(discount)
```

```python
#To make any changes to global variable mention it as global inside the function
#Without mention of global function thinks you are referring to the local variable
discount=13 #--> Global variable


def get_discount():
    global discount #use global keyword to make changes to it inside the function
    discount+=1 #--> Local variable
    print("Discoun added:",discount,"%")


get_discount()
print(discount)
```

```python
# BUILT-IN — Python's own names like print, len, sum
# available everywhere without defining them

# GLOBAL — defined at the top level of the file
store_name   = "QuickCart"
tax_rate     = 0.18

def apply_offer(order_amount):

    # LOCAL — defined inside this function
    # only visible inside apply_offer
    offer_discount = 10

    def calculate_final():

        # ENCLOSING — calculate_final can see
        # order_amount and offer_discount from apply_offer
        # without them being defined here

        subtotal    = order_amount - (order_amount * offer_discount / 100)
        tax         = subtotal * tax_rate           # tax_rate from GLOBAL
        final       = subtotal + tax

        # BUILT-IN — len() and print() used here
        # neither defined locally nor passed in
        store_label = store_name.upper()            # store_name from GLOBAL
        print(f"Store: {store_label}")              # print — BUILT-IN
        print(f"Items in label: {len(store_label)}")# len   — BUILT-IN
        print(f"Subtotal:  Rs.{subtotal}")
        print(f"Tax (18%): Rs.{tax:.2f}")
        print(f"Final:     Rs.{final:.2f}")

        return final

    return calculate_final()

result = apply_offer(2000)
```

* Looking for store_name:
  * L — defined in calculate_final?  No
  * E — defined in apply_offer?      No
  * G — defined at top of file?      Yes → "QuickCart"

* Looking for order_amount:
  * L — defined in calculate_final?  No
  * E — defined in apply_offer?      Yes → 2000

* Looking for offer_discount:
  * L — defined in calculate_final?  No
  * E — defined in apply_offer?      Yes → 10

* Looking for tax_rate:
  * L — defined in calculate_final?  No
  * E — defined in apply_offer?      No
  * G — defined at top of file?      Yes → 0.18

* Looking for print:
  * L — No
  * E — No
  * G — No
  * B — Yes → Python built-in

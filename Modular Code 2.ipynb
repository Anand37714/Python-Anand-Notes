# *args --> Rules and Properties

1. *args collects all extra positional arguments into a tuple.
2. The name args is convention --> you can use any name after *.
3. *args must come after regular positional parameters.
4. you can loop over *args --> it is tuple.
5. only one *args can be used per function

```python
#Function definiyion with args parameter
#Used when we don't know how many positional arguments will be passed
def total_bill(*bill_amounts):
    print(type(bill_amounts))
    print(bill_amounts)
    return sum(bill_amounts)

print(total_bill(670,780,450,700))
print(total_bill(12,45,90))
```

```python
def create_order(status,*order_items):
    print(type(order_items))
    print(status)
    for order in order_items:
        print("order:",order)
    return "orders printed"

print(create_order("delivered","shoe","bag","bottle"))
print(create_order("pending","jersey","pen","paint","keyboard","mouse"))
```

```python
#create a function taking multiple order_ids, print the type of argument
#also print the order id one by one using loop (take atleast 5 arguments)
#output -   Processing:ORD001
#           Processing:ORD002

def orders(*order_ids):
    print(type(order_ids))
    for i in order_ids:
        print("Processing:",i)
    return "Orders Processed"

print(orders("ORD001","ORD002","ORD003","ORD004","ORD005"))
print(orders("ORD006","ORD007","ORD008","ORD009","ORD0010"))
```

```python
#Take multiple order amounts and print the total along with order amounts

#Output:
#   orders:(1500,700)
#   Total:Rs.2200
#   orders:(1500,)
#   Total:Rs.1500

def orders(*bill_amounts):
    print("orders:",bill_amounts)
    print("Total: Rs.",sum(bill_amounts))
    return "orders closed"

print(orders(1500,700))
print(orders(1500,))
```

## **Kwargs --> Let you pass any number of keyword arguments. pyhton collects them into dictionary
### Rules and Properties:
1. **kwargs collects all extra keyword arguments into a dictionary
2. The name kwargs is convention --> you can use any name after **
3. **kwargs must come after regular parameters and *args
4. keys are strings --> the argument names become dictionary keys
5. only one **kwargs can be used per function

```python
# *Kwargs (KW=key word)
def order_details(**ord):
    print(type(ord))
    for k,v in ord.items():
        print(f"{k}:{v}")

order_details(product="Bottle",quantity="7",stock=10,color="red")
```

```python
def create_order(order_id,*items,**details):
    print("Order:",order_id)
    print("Items:",items)
    print("Details:",details)
    return "Items Details"
print(create_order("ORD001","Shoe","bag","bottle",city="Banglore",status="Pending",payment="card"))
print(create_order("ORD002","Shoe","bag","pen",city="Manglore",status="pending",payment="upi"))
```

# Rules and Properties of Lambda
1. Written in one line --> cannot span multiple lines'
2. No def, No return required here --> the expression is automatically returned
3. can have any nyumber of parameters-including none type also
4. only one expression --> cannot contain statements, loops, or conditions
5. often used inline-passed directly to another function

```python
#Lambda --> Lambda is an anonymous function without a name, written in one line.

#Regular function:
def double(x):
    return x*2

print(double(2))

#Lambda function
#Syntax --> lambda parameter:expression 
double = lambda x:x*2
Square = lambda y:y**2
print(double(2))
print(Square(5))

```

```python
#Basic lambda examples:

#with no parameter
greet = lambda:"hello"
print(greet())

#with one parameter
cube = lambda z:z**3
print(cube(5))

#with two or more parameters
discount = lambda amount,discount : amount-(amount*discount)
print(discount(400,0.05)) #positional arguments

#with two or more parameters for keyword arguments
order_details = lambda product,quantity,amount : f"The quantity purchase is:{quantity}, total cost:{amount}, product purchased:{product}"
print(order_details(quantity=4,product="Markers",amount=400)) #Keyword arguments
```

```python
#with condition
print("with condition")
get_delivery = lambda amount:"free delivery" if amount>1000 else "Rs.50/- will be charged"
print(get_delivery(500))
print(get_delivery(1500))

#with default value
get_delivery = lambda amount,threshold=1000:"free delivery" if amount>threshold else "Rs.50/- will be charged"
print("with default value")
print(get_delivery(1500))
print(get_delivery(500))

#with args
print("with args")
order_total = lambda *order_amounts:sum(order_amounts)
print(order_total(500,700,800))
print(order_total(500,700,800,900,1000))

#with kwargs
print("with kwargs")
order_details = lambda **details:f"The product details are:{details}"
print(order_details(product="Marker",amount=50,quantity=1,city="Banglore"))
```

# Dictionaries 
* Values are stored as Key:Value pairs
* Ordered
* Duplicates key are not allowed --> Means  Only immutable elements can be taken as key values
* Heterogeneous
* Mutable / Changable
* Keys must be immutable elements, Values can be mutable elements

```python
#Definition
Fruits = {"Apple":"Red", "Banana":"Yellow", "Orange":"Orange", "Grapes":"Green"}
print(Fruits)
print(type(Fruits))

Students = ["Kusuma", "Ganesh", "Narmada", "Sanjana"]
classes = ["Class1", "Class2", "Class3", "Class4",]
d = dict(zip(Students,classes))
print(d)
print(type(d))

Students = ["Kusuma", "Ganesh", "Narmada", "Sanjana"]
classes = ["Class1", "Class2"]
d = dict(zip(Students,classes))
print(d)
print(type(d))
```

```python
#Accessing the values
#Accessing the single value
Fruits = {"Apple":"Red", "Banana":"Yellow", "Orange":"Orange", "Grapes":"Green"}
print(Fruits["Apple"])

#Keys and Values are as different data types : Heterogeneous Property
Fruits = {1:"Red", 2:"Yellow", 3:"Orange", 4:"Green"}
print(Fruits)
print(Fruits[2])

#Heterogeneous Property
Fruits = {"Apple":1, "Banana":1.5, "Orange":True, "Grapes":"Green"}
print(Fruits["Orange"])

#Lists, Sets, dicts are mutable and cannot be stored as keys but tuples can be
Fruits = {1:"Red", 1.5:"Yellow", True:"Orange", "Grapes":"Green", (1,2,3):"Rainbow"}
print(Fruits[(1,2,3)])
print(Fruits[True])
print(Fruits)

#Duplicate key behaviour
Fruits = {"Apple":1, "Apple":1.5, "Orange":True, "Grapes":"Green"}
print(Fruits)
```

```python
#Accessing a value through Get method
Fruits = {"Apple":"Red", "Banana":"Yellow", "Orange":"Orange", "Grapes":"Green"}
print(Fruits.get("Apple")) #Access a value

print(Fruits.get("Pear")) 
#Safe method to access a value
#Get method does not throw a key aerror if the given value is not present

print(Fruits.get("Pear", "N/A"))
#CUSTOM MESSAGE IF KEY NOT FOUND
```

```python
#Update, pop, copy, clear

#Update
Fruits = {"Apple":"Red", "Banana":"Yellow", "Orange":"Orange", "Grapes":"Green"}
Fruits["Apple"] = "Blue" #Updating a value through key
print(Fruits)

Fruits.update({"Apple":"Pink"}) #Updating single element
print(Fruits)

Fruits.update({"Apple":"Pink","Banana":"Blue"}) #Updating multiple elements
print(Fruits)

#There is no specific Insert method here in dictionaries
#If the key is present and give the new key it will replace with the new value
#If the key:value pair is not present and give the new key:value pair it will add

#Remove
Fruits.pop("Apple") #Removing an item through key
print(Fruits)

#Fruits.pop("Peach"}) --> if you try to remove an element using pop if the value is not present there it will throw an error


Fruits.pop("Peach","not found") #Custom messasge if the value not found
#without custom message pop throws error if key not found
```

```python
#copy
backup = Fruits.copy()
print(Fruits)

#Clear --> Clears the entire dictionary
Fruits.clear()
print(Fruits)
```

```python
doctor = {
    "id"         : "D001",
    "name"       : "Dr. Sharma",
    "department" : "Cardiology",
    "experience" : 15,
    "salary"     : 120000
}

# 1. Print the doctor's name
print("doctor name :", doctor["name"])

# 2. Print the department using direct access
print("deparment :", doctor["department"])

# 3. Print the salary using .get()
print("salary :", doctor.get("salary")) # ---> Doubt here with using type brackets

# 4. Try accessing "age" directly — what happens?
try : print("age :", doctor["age"])
except  KeyError : print("key error : age does not exist")  

# 5. Access "age" using .get() with default "Not recorded"
print("age :", doctor.get("age","Not recorded"))  # ---> Doubt here with using type brackets

# 6. Add a new key "patients_today" with value 12
doctor["patients_today"] = 12
print(doctor)

# 7. Update salary to 135000
doctor["salary"] = 135000
print("After salary update :", doctor)

# 8. Update both experience and salary at the same time using .update()
doctor.update({"experience":20, "salary":140000})
print("After experience and salary update :", doctor)

# 9. Delete the "id" key using del
del doctor["id"]
print("After delete:", doctor)

# 10. Print all keys as a list
keys = list(doctor.keys())
print("keys :", keys)
print("keys :", list(doctor.keys()))

# 11. Print all values as a list
values = list(doctor.values())
print("values :", values)
print("values :", list(doctor.values()))

# 12. Print all key-value pairs using .items()
Items = list(doctor.items())
print("Items :", Items)
print("Items :", list(doctor.items()))

# 13. Remove "department" using .pop() — print what was removed
Removed = doctor.pop("department")
print("Removed :", Removed)
print("After pop :", doctor)

# 14. Make a copy called backup_doctor
backup_doctor = doctor.copy()
print("Back up doctor :", backup_doctor)

# 15. Clear the original — print both
doctor.clear()
print("After clear :", doctor)
print("Back up doctor :", backup_doctor)
```

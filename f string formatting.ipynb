```python
name = "Anand"
age = 30
# Normal method
print("The name is", name, "and the age is", age)

# f-string formatting --> It is a method of formatting the output
name = "Anand"
age = 30
print(f"The name is {name} and the age is {age}")
```

```python
# f-string formatting can be used for three things
# Allignment
# Padding
# Precisioning

### Allignment Method ###
print(f"|{'Anand':>10}|") #Right allign
print(f"|{'Anand':<10}|") #Left allign
print(f"|{'Anand':^10}|") #Centre allign
print(f"|{'Mahesh':10}|") #Alligns texts by default to the left when < or > not mentioned
print(f"|{'Mahesh':>10}|")
print(f"|{101:10}|") ##Alligns Numbers by default to the right when < or > not mentioned
print(f"|{101:<10}|")
### Padding Method ### ---> Filling the characters in blank space
print(f"|{'Gadag':10}|") #Alligns text by default to the left when < or > not mentioned

print(f"|{'Mahesh':010}|") #0 is filled in the place of blank space
print(f"|{'Mahesh':*<10}|") #* is filled after the name Mahesh from left side in the place of blank space
print(f"|{'Mahesh':*>10}|") #* is filled after the name Mahesh from right side in the place of blank space
print(f"|{'Anand':*^10}|") #* is filled in the blank space at both the side of the name Anand

print(f"|{100:*<10}|")
print(f"|{100:*>10}|")
print(f"|{100:*^10}|")

#Note that in centre allignment when splitting
#Between number of characters to fill in the left and right side with odd numbers
#It fills 1 less on left side and 1 more on right side

### Precisioning Method ### ---> Upto how many decimal places should float values be formatted
#.nf --> Upto nth decimal places
print(f"{12345.78964123:.0f}")
print(f"{12345.78964123:.1f}")
print(f"{12345.78964123:.2f}")
print(f"{12345.78964123:.3f}")
print(f"{12345.78964123:.4f}")
print(f"{12345.78964123:.5f}")

a=45686.489654123
print(f"{a:.0f}")
print(f"{a:.1f}")
print(f"{a:.2f}")
print(f"{a:.3f}")
print(f"{a:.4f}")
print(f"{a:.5f}")

```

```python
# Padding and Precision together

print(f"{854.2617:010.1f}") # For integer the allignment is Right side by default
print(f"{854.2617:<010.1f}")
print(f"{854.2617:>010.1f}")
print(f"{854.2617:^010.1f}")
#print(f"{"Anand":010.1f}") # For Characters/Strings the Padding and Precision together will not work Because of Unknown format code 'f' for object of type 'str'
```

```python
# GENERATE THESE INVOICE NUMBERS IN THIS FORMAT "INV-0001, INV-0002, INV-0003"
print(f"INV-{1:04}") # Numbers/Integers are alligned right side by default
print(f"INV-{2:04}") # Numbers/Integers are alligned right side by default
print(f"INV-{3:04}") # Numbers/Integers are alligned right side by default
```

Advanced f-string formatting

{} - to use the variable value between printing

.nf - n number of decimal places

<   - left-align

>   - right-align

^  - center align

,  - thousands separator

b - binary

o - octal

% - formatting the value as percentage value

.g - Remove trailing zeroes

```python
Salary = 1250000
print(f"His salary is {Salary:,}")

a=255
print(f"Binary value is {a:b}")
print(f"Octal value is {a:o}")

b=123687.3786898500000
print(f"The value of b is {b:}")
print(f"Round of the value of b is {b:.3f}")
print(f"The value of b is {b:%}")
print(f"The value of b is {b:.3%}")
```

```python
#A shop gives a discount of 12.5% on 4378.
#Print the discount amount and final price
#Both to 2 decimal places with thousands seperator

price = 4377
discount = price*0.125
final_amount = price - discount

print(f"The discounted amount is {discount:.2f}")
print(f"The final price is {final_amount:,.2f}")

#Another method
price = 4377
discount = 0.125

print(f"The discounted amount of (4378) with {discount:.2%} is {(price*discount):.2f}")
print(f"The final price after discount is {price-(price*discount):.2f}")

```

Rules and Properties of Numpy-

1. All elements must be the same data type-unlike lists which can mix strings,ints,booleans.
2. Size is fixed after creation-you cannot append to a Numpy array like a list. You have to create a new one.
3. Operations apply to every element automatically-no loops needed
4. Faster and more memory efficient than Python lists -especially for large data.
5. Shape describes dimensions -(5,) means 1D with 5 elements. (3,4) means 2D with 3 rows and 4 columns

```python
#Vectorization
#Vectorization means performing an operation on an entire array at once instead of looping through each element one by one 

prices=[85000,25000,45000,5000,15000]

#without vectorization - loop is needed
discounted=[]
for i in prices:
    discounted.append(i*0.9)
print(discounted)

#with vectorization
prices=np.array([85000,25000,45000,5000,15000])
discounted=prices*0.9
print(discounted)
```

```python
#Why Vectorization is faster?
#python loops are interpreted line by line which is slow
#numpy skips the python interpreter for the actual math

import time
import numpy as np

#1 million prices
prices=np.array([100]*1000000)

#LOOP APPROACH
start=time.time()
result=[p*0.9 for p in prices]
end=time.time()
print(f"Loop: {end-start:.4f} Seconds")

#Vectorized approach
start=time.time()
result=prices*0.9
end=time.time()
print(f"Vectorized: {end-start:.4f} Seconds")

#Vectorized approach is roughly 87 times faster, the gap grows as data grows
```

### What is Broadcasting?
* Broadcasting is NumPy's ability to perform operations on arrays of different shapes — by automatically expanding the smaller array to match the larger one.
* Without broadcasting you could only do math between arrays of identical shapes. 
* Broadcasting removes that restriction.

```python
#Scalar-single value based operation
prices=np.array([85000,25000,45000,5000,15000])
discount=0.9
#discount=[0.9,0.9,0.9,0.9,0.9] - internally it streches discount then multiply element-wise

result=prices*discount
print(result)
```

### Broadcasting Rules
* Python follows these three rules in order:
* Rule 1 — If arrays have different number of dimensions, the shape of the smaller one is padded with 1s on the left side.
* Rule 2 — Dimensions with size 1 are stretched to match the other array along that dimension.
* Rule 3 — If dimensions do not match and neither is 1 — ValueError. Cannot broadcas

```python
#Broadcasting 1D array with 2D array

#Monthly sales- 3 products X 3 months

sales=np.array([[120,135,110], #laptop
                [200,180,220], #mobile
                [80,95,75]]) #tablet
print(sales.shape)

#Apply different growth rate to each month
growth=np.array([1.1,1.2,1.05])
print(growth.shape)

#Broadcasting -sales*growth
adjusted=sales*growth
print(adjusted)
```

3D Arrays: A 3D array is an array of 2D arrays. Think of it as a multiple tables stacked
| Dimensions | Think of it as |
|---|---|
| 1D | A single row — one product's prices |
| 2D | A table — products × months |
| 3D | Multiple tables — products × months × years |

```python
import numpy as np

#sales data
#2 years, 3 products, 4 months

#Year 2023

year_2023=[[120,135,110,145], #laptop
           [200,180,220,190], #mobile
           [80,95,75,100]] #tablet
            #J  F  M  A

year_2024=[[150,160,140,170], #laptop
           [220,210,240,230], #mobilE
           [90,105,85,115]] #tablet
            #J  F  M  A

#Stak them into a 3D array
sales=np.array([year_2023,year_2024])

print(sales)
print(sales.shape) #(2,3,4) - 2 tables,3 rows,4 columns
print(sales.ndim) #3 - 3D
print(sales.size) #24 - total elements
```

```python
# 3D Dimension using Numpy style by mention the 2tables in a single bracket
sales=np.array([
                [
                [120,135,110,145],
                [200,180,220,190],
                [80,95,75,100]
                ],
            
                [
                [150,160,140,170], 
                [220,210,240,230],
                [90,105,85,115]
                ]
                ])
print(sales)
print(sales.shape)
print(sales.ndim)
```

| Axis | Size | Represents |
|---|---|---|
| Axis 0 | 2 | Years |
| Axis 1 | 3 | Products |
| Axis 2 | 4 | Months |

```python
#Indexing a 3D array
#Syntax:[Table,row,column]
#       [Year,Product,Month]

#Year 2023-entire table
print(sales[0])

#Year 2024-entire table
print(sales[1])

#Year 2023-laptop row
print(sales[0,0])

#Year 2024-mobile row
print(sales[1,1])

#Year 2023-laptop, February
print(sales[0,0,1])

#Year 2024-tablet, march
print(sales[1,2,2])
```

```python
#Slicing a 3D array

#all years, laptop only, all months
print(sales[:,0,:])

#all years,all products,jan only
print(sales[:,:,0])

#year 2023,first 2 products,first 2 months
print(sales[0,:2,:2])
```

```python
#Operations on 3D arrays

#Apply 10% growth to all values
import numpy as np
print(sales*1.1)

#Total sales across everything
print(np.sum(sales))

#Total per year-sum across products and months
print(np.sum(sales,axis=(1,2)))

#Average per product across all years and months
print(np.average(sales,axis=(0,2)))
```

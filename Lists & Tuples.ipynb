```python
# check whether behaviour changes with inner list
b=[10,2.5,67,[4,5,6]]
print(b)

c=b.copy()
print(c)
print(b)

b[0]=3
print(b) # change made to b

c[0]=5
print(c) # change made to c
print(b) # change made to normal elements of c does not affect b



```

```python
#Behaviour changes when there is an inner list
b=[10,2.5,67,[1,2,3]]
print(b)

c=b.copy()
print(c)

b[3][0]=10
print(b)

c[3][0]=20
print(c)
print(b) # change made to c is also affecting b (bcz it was only a shallow copy)

#copy method only creates a shallow copyA
```

```python
# we use the word import when we want something that is not 
import copy
b=[10,2.5,67,[1,2,3]]
c=copy.deepcopy(b) #deep copy is part of the copy module 
print(c)

b[3][0]=10
print(b)

c[3][0]=20
print(c)
print(b)
```

```python
# Tuples
#A Tuple is ordered, immutable, indexed, allows duplicates, heterogeneous
#creating a tuple

a=(1,2,3,3,3,'hey',3.5)
print(type(a))

b=1,2,3         #packing a tuple
print(b)
print(type(b))

c=[4,5,6] #packing a list
print(c)
print(type(c))

e=tuple(c)
print(e)
print(type(e))

d=tuple([4,5,6])
print(d)
print(type(d))

f=(5,) # creating a single element tuple
print(f)
print(type(f))


```

```python
# Packing and Unpacking a tuple

b=1,2,3
print(b)
print(type(b))

a,b,c=(1,2,4)    # Unpacking a tuple
print(a)
print(b)
print(c)

g=(1,2,6)
print(type(g))

e,f,c=g  # Unpacking a tuple through a variable 'g' that has tuple stored in it
print(e)
print(f)
print(c)
```

```python
# Extended unpacking
# Unpacking multiple elements into fewer variables
*a,b,c=(1,2,3,4,5,6)
print(a)
print(b)
print(c)

a,*b,c=(1,2,3,4,5,6)
print(a)
print(b)
print(c)

a,b,*c=(1,2,3,4,5,6)
print(a)
print(b)
print(c)
```

```python
# Tuple repitition
t=(1,)*3
print(t)
print(type(t))

#list repitition
l2 = [3]*3
print(l2)
print(type(l2))

# Tuple concatenation
t1 = (1,2,3)
t2 = (4,5,6)
t3=t1+t2
print(t3)
print(type(t3))
```

```python
#Creating the empty tuple
t=()
print(t)
print(type(t))


#Creating Empty list
l=[]
print(l)
print(type(l))
```

```python
t1=(1,1,3,4.6,'Hey','True')

print(t1[0])   #accessing

print(t1[:])  #slicing  #[start:stop:step]

print(t1[::-1])   #negative index #reverse tuple

print(t1[::2])   #every second element through step
```

```python
#Tuple methods

t2=(1,2,3,4,5,6)
print(sum(t2))

t3=(1,5,8,10,0,4)
print(sorted(t3)) #Sorting in Ascending order
print(sorted(t3,reverse=True))

print(max(t3))
print(min(t3))

#specific to tuples -tuple methods

#count
#index

t2=(2,2,3,6,5,2,6,8,5,6,3,4,5)
print(t2.count(6))   #counts the number of occurences of an element

print(t2.index(5,10))   #index of a value along with where to count from or start value
#here 5 means find the position of 5 after skipping the 10 values (position count starts from 0)

del t2  #can only delete entire tuple
```

```python
#Tuple comparision
t1=(1,2,3,4,5)
t2=(1,2,3,4,5)

print(t1==t2)
print(t1<t2)
print(t1>t2)
```

```python
#Tuple  itself is immutable but it can store multiple elements
t1=(1,2,3.4,'hey',[1,2,3])
t1[4]

t1[4][0]=5 #assigning the new value into a tuple
print(t1)

t1[4].append(7)
print(t1)

#Tuple conversion
t5=(2,3,4,5)
print(type(t5))
t6=tuple(t5)
print(t6)
print(type(t6))

t7=list(t6)
print(t7)
print(type(t7))
```

Advantages of Tuples-

* Faster compared to list
* Memory efficient - Consumes less memory
* Immutable - Prevents accidental modifications
* Hashable - can be dictionary keys

When to use Tuples-

* Data should not change
* To store coordinates
* Database records
* Dictionary keys
* Fixed collections

Tuple methods and Return types
| Method | What it does | Returns |
|---|---|---|
| `.count()` | How many times value appears | Integer |
| `.index()` | Position of first occurrence | Integer |

When to use tuples and when to use lists? 
| Use List | Use Tuple |
|---|---|
| Patient queue — keeps changing | Admission record — fixed |
| Bed availability — updates daily | Transaction ID + date — sealed |
| Staff roster — add/remove | Doctor ID + specialization — fixed |

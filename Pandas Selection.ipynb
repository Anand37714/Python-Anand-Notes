## Pandas Selection
### What is selection ?
Selection means retrieving specific rows, columns, or individual values from a DataFrame.

In analysis, we almost never work with an entire dataset at once. Instead we extract only the information required to answer a business question.

Example:
* Show all players from spain.
* Display only the player name and goals
* Find players who scored more than 2 goals

```python
#Setup- Load the dataset

import pandas as pd
#cols=["player_name","nationality","age","player_id","goals","match_date","position","player_rating"]
df=pd.read_csv("./datasets/fifa_world_cup_2026_player_performance.csv",parse_dates=["match_date"])

df["match_date"]=pd.to_datetime(df["match_date"],format="%d-%m-%Y")

#check the size of the dataset
print(df.shape)
df.head()
```

## Four Ways to Select Data

| Method | Description | Use When |
|--------|-------------|----------|
| `[]` | Select one or more columns | You only need columns |
| `.loc[]` | Label-based selection | You know the row labels or column names |
| `.iloc[]` | Position-based selection | You know the row or column positions |
| Boolean Masking / `.query()` | Filter rows using conditions | You want records matching a condition |

```python
#Selecting Columns

#The simplest form of selection is selecting one or more columns

#Selecting a single column
#Syntax: df["column_name"]

player_names=df["player_name"]
print(player_names.head())
print(type(player_names))

#Selecting a single column returns a series,a 1D object,think of it as a single column extracted from a table
```

```python
#Selecting multiple columns
#Syntax: df[["column1","column2"]]

#Notice the double square brackets

#Outer brackets-tell pandas you're selecting columns'
#Inner brackets-creates a python list containing multiple column names

players = df[[
    "player_name",
    "nationality",
    "position",
    "goals",
    "player_rating"
]]

print(players.head())
print(type(players))
```

Use a Series when:
* You need values from one column.
* You want to perform calculations.
* You want to use methods like:

df["goals"].mean()

df["goals"].sum()

df["player_rating"].max()

Use a DataFrame when:
* You want to keep the tabular structure.
* You plan to add more columns later.
* You want to merge or join with another DataFrame.
* You are passing data to functions that expect a DataFrame.

```python
#.loc[]- Label baed selection

#.loc[] is used to select rows and columns using labels

#Labels refers to:
#Roe-index labels
#Column names

#Syntax: df.loc[row_label,column_label]

#Select a single row

print(df.loc[0])

#Select multiple rows
print(df.loc[[0,1,2,3,4]])
```

### i.loc - position based selection
* iloc[ ] is used to select rows and columns using integer positions.
* Instead of row labels and column names, .iloc[ ] uses the position of rows and columns.
* Row and column positions always start from 0.

### Syntax - df.iloc[row_position, column_position]

### Important Rules
* .iloc[ ] uses positions, not labels.
* Positions always start from 0.
* When slicing, the ending position is NOT included.
* Works exactly like Python list slicing.

| Selection | Syntax |
|-----------|--------|
| First Row | `df.iloc[0]` |
| Multiple Rows | `df.iloc[[0,5,10]]` |
| Row Range | `df.iloc[0:5]` |
| Single Column | `df.iloc[:,1]` |
| Multiple Columns | `df.iloc[:, [1,4,22]]` |
| Rows & Columns | `df.iloc[0:5,[1,4,22]]` |
| Single Value | `df.iloc[2,1]` |
| Last Row | `df.iloc[-1]` |

```python
#Select first row

import pandas as pd
df=pd.read_csv("./datasets/fifa_world_cup_2026_player_performance.csv",parse_dates=["match_date"])
df["match_date"]=pd.to_datetime(df["match_date"],format="%d-%m-%Y")


#check the size of the dataset
print(df.shape)
df.head()

print(df.iloc[[0]]) #single row
print()
print(df.iloc[[0,5,10]]) #multiple rows
print()
print(df.iloc[0:5]) #range of rows
print()
print(df.iloc[:,1]) #single column
print()
print(df.iloc[:, [1,4,22]]) #multiple column

```

```python
#Boolean masking

#Boolean masking is used to filter rows based on one or more conditions.
#Instead of selecting rows by their labels or positions, we select rows that satisfy a condition.

#For example:
#1. Display players who scored more than 2 goals
#2. Show all spanish players
#3. Find players with a rating greater than 8
#4. Display goalkeepers
```

```python
#Boolean masking is done in two steps

#Step1: Create a boolean mask

mask=df["goals"]>0
print(mask.head())

#Step2: Apply the boolean mask
scorers=df[mask]
print(scorers.head())

#printing as mask directly
mask=df["goals"]>0
print(df[mask].head())

#Even more direct method
print(df[df["goals"]>0].head())
```

```python
#Players younger than 21
print(df[df["age"]<21].head())

#Spanish players
print(df[df["nationality"]=="Spanish"].head())

#Players who are not goal keepers
print(df[df["position"]!="goalkeeper"].head())
print(df[~(df["position"]=="goalkeeper")].head())

#Multiple Conditions
#Players who are spanish and have goals>0
print(df[(df["nationality"]=="Spanish") & (df["goals"]>0)])

#Players who have either goals>0 or assists>0
print(df[(df["assists"]>0) | (df["goals"]>0)])

#Players who are spanish, brazzilian or french
print(df[(df["nationality"]=="Spanish") | (df["nationality"]=="Brazilian") | (df["nationality"]=="French")])
print(df[df["nationality"].isin(["Spanish","Brazilian","French"])])
```

```python
#Selecting columns after filtering
#Display columns --> player name, goals, player rating whose goals>0
print(df.loc[df["goals"]>0,["player_name","goals","player_rating"]])
```

```python
#More examples on boolean masking with .loc
# Spanish players who scored — show name, goals, rating
df.loc[
    (df["nationality"] == "Spanish") & (df["goals"] > 0),
    ["player_name", "goals", "player_rating"]
].head()

# Players rated above 8 who played 90 minutes — show name, position, rating
df.loc[
    (df["player_rating"] >= 8.0) & (df["minutes_played"] == 90),
    ["player_name", "position", "player_rating", "minutes_played"]
].head()

# Forwards who scored more than 1 goal — show name, goals, assists
df.loc[
    (df["position"] == "Forward") & (df["goals"] > 1),
    ["player_name", "goals", "assists"]
].head()

# Players under 21 — show name, age, nationality, position
df.loc[
    df["age"] < 21,
    ["player_name", "age", "nationality", "position"]
].head()

# Players with yellow card — show name, nationality, yellow_cards
df.loc[
    df["yellow_cards"] > 0,
    ["player_name", "nationality", "yellow_cards"]
].head()
```

```python
#Boolean Mask with .iloc[]

#.iloc[] only accepts integer positions — not column names. So you can use a boolean mask for rows but must use integer positions for columns.
```

```python
# Boolean mask for rows — integer positions for columns
mask   = df["goals"] > 0
result = df.iloc[mask.values, [1, 3, 27, 61]]   # columns by position
#or
result = df.iloc[(df["goals"]>0).values, [1, 3, 27, 61]]   # columns by position
print(result.head())
```

```python
# Example 1 — Single condition, multiple columns
# Players who got a red card
df.loc[
    df["red_cards"] > 0,
    ["player_name", "nationality", "position", "red_cards"]
]

# Example 2 — Two conditions, range of columns
# Spanish midfielders
df.loc[
    (df["nationality"] == "Spanish") & (df["position"] == "Midfielder"),
    "player_name":"position"    # column range
]

# Example 3 — Three conditions, specific columns
# Young players under 21 who scored and got high rating
df.loc[
    (df["age"] < 21) & (df["goals"] > 0) & (df["player_rating"] >= 7.0),
    ["player_name", "age", "goals", "player_rating"]
]

# Example 4 — OR condition with columns
# Players who scored OR assisted in Final stage
df.loc[
    ((df["goals"] > 0) | (df["assists"] > 0)) & (df["tournament_stage"] == "Final"),
    ["player_name", "tournament_stage", "goals", "assists"]
]

# Example 5 — NOT condition with columns
# Non-goalkeepers who played full 90 minutes
df.loc[
    (~(df["position"] == "Goalkeeper")) & (df["minutes_played"] == 90),
    ["player_name", "position", "minutes_played", "player_rating"]
]

# Example 6 — isin() with columns
# Players from top nations in knockout stages
knockout_stages = ["Quarter-Final", "Semi-Final", "Final"]
top_nations     = ["Spanish", "Brazilian", "French", "German"]

df.loc[
    (df["tournament_stage"].isin(knockout_stages)) &
    (df["nationality"].isin(top_nations)),
    ["player_name", "nationality", "tournament_stage", "goals"]
]

# Example 7 — str.contains() with columns
# Players with "Gavi" in name — show all performance columns
df.loc[
    df["player_name"].str.contains("Gavi"),
    ["player_name", "nationality", "goals", "assists", "player_rating"]
]

# Example 8 — Multiple conditions with column range
# High rated players in Group Stage
df.loc[
    (df["player_rating"] >= 8.5) & (df["tournament_stage"] == "Group Stage"),
    "player_name":"position"
]
```

```python
#Querying
#.query()- string based filtering

#.query is a clean way to filter rows using a string expression instead of boolean masking.
#Easier to read especially for multiple conditions

#Syntax: df.query("condition")

#Goals greater than zero
df.query("goals>0")

#Age less than or equal to 21
df.query("age<=21")

#Specific nationality
df.query("nationality=='Spanish'")
```

```python
#goals>0 and nationality is spanish
df.query("goals>0 & nationality=='Spanish'")
df.query("goals>0 and nationality=='Spanish'")

#goals>0 and assists>0
df.query("goals>0 | assists>0")
df.query("goals>0 or assists>0")

#Combined
df.query("goals>0 and assists>0 and player_rating>=7.0")
```

```python
#Query with variables

min_rating=8.0
nationality="Spanish"

df.query("player_rating>=@min_rating and nationality==@nationality")

#@ Symbol tells the query that this is a python variable from outside and not a column name
```

```python
#Boolean masking V/S .query()

#using .query()
#Combined
df.query("goals>0 and assists>0 and player_rating>=7.0")

#using boolean masking
df[(df["goals"]>0) & (df["assists"]>0) & (df["player_rating"]>=7.0)]

#boolean masking with variable
rating=8.0
df[(df["goals"]>0) & (df["assists"]>0) & (df["player_rating"]>=rating)]
```

```python
#if column name has spaces then use backticks in .query()
df.query("`jersey number`>3")
```

```python
# .query with .loc
#display only player_name, goals, player_rating
df.query("goals>2").loc[:,["player_name","goals","player_rating"]]

# .query with .iloc
#nationality is brazilian and select rows from 0 to 4
df.query("nationality=='Brazilian'").iloc[:5]
```
